# Conjuntos

---

Las listas tiene el operador `in` que permite comprobar si un elemento pertenece a una lista:

```python
lst = [1, 5, 7, 9]
5 in lst, 2 in lst
```

Cuando las listas son grandes el operador `in` es ineficiente puesto que no tiene ningún criterio para saber la posible posición en la que se encuentra el elemento buscado. 

Si la posición en la que se encuentran los elementos no es importante, podemos usar conjuntos para que las búsquedas sean eficientes.

```python
st = {'pepe', 'juan', 'luis', 'alfredo', 'luis'}
'pepe' in st
```

```python
'John' in st
```

Los **conjuntos** son estructuras de elementos que cumplen
* El orden no es relevante. Se puenden recorrer como las listas, pero no tenemos ninguna garantía del orden en el que salen los elementos.
* No hay repetición de elementos.
* Pueden contener elementos que tengan definida una función *hash*. En principio elementos inmutables. Los conjuntos pueden contener tuplas pero no listas 

```python
for x in st:
    print(x)
```

```python
st2 = {(1,2), (1,4), (2,5)}
st2
```

## Operaciones

Tienen las operaciones típicas de los conjuntos matemáticos: unión, intersección, diferencia y diferencia simétrica.


```python
conj_a = {"a", "e", "i", "o", "u"}
conj_b = {"a", "b", "r", "a", "c", "a", "d", "a", "b", "r", "a"}
conj_c = {"a", "o", "u" }
# Y con ellos es posible usar las operaciones típicas de conjuntos:

print(f"A = {conj_a}")
print(f"B = {conj_b}")
print(f"A unión B = {conj_a | conj_b}")   
print(f"A intersección B = {conj_a & conj_b}") 
print(f"A diferencia B = { conj_a - conj_b}")
print(f"A dif. simétrica B = { conj_a ^ conj_b}")
print(f"¿{conj_c} es subconjuno de A? {conj_c < conj_a}")
print(f"¿{conj_c} es subconjuno de B? {conj_c < conj_b}")
```

Las operaciones anteriores crean objetos nuevos. Existen versiones con métodos que también devuelven objetos nuevos.

```python

conj_a = {"a", "e", "i", "o", "u"}
conj_b = {"a", "b", "r", "a", "c", "a", "d", "a", "b", "r", "a"}
conj_c = {"a", "o", "u" }

print("A = ", conj_a)
print("B = ", conj_b)
print(conj_a.union(conj_b))
print(conj_a.intersection({"a", "r", "p", "e", "g", "i", "o"}))
print(conj_a.difference({"a", "g", "r", "i", "o"}))
print("A = ", conj_a)
print("B = ", conj_b)
print(conj_c.issubset(conj_a))
print(conj_c.issubset(conj_b))
```

También se pueden generar conjuntos intensionales

```python
pairs = { (x, y) for x in range(3) for y in range(3) if x!=y }
pairs
```
