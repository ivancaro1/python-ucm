## El lenguaje Python

__Python es un lenguaje de propósito general__ (con el que se pueden desarrollar aplicaciones muy diversas) que permite escribir código compacto y sencillo de leer.

Su creador fue Guido Van Rossum (1990) y debido a que es de código libre la comunidad de desarrolladores han creado librerías o módulos para hacer casi cualquier cosa. Se trata de un lenguaje que permite desarrollar un proyecto completo o puede ser parte de un sistema.

Se recomienda aprender a programar Python, por cosas muy sencillas:

* Su filosofía de diseño hace hincapié en escribir código legible y sencillo de entender, y por ese motivo se usa en muchos centros de enseñanza para enseñar a programar.
* La curva de aprendizaje es muy corta y en muy poco tiempo uno puede estar programando cosas no tan triviales.
* Es fácil de aprender, fácil de leer y cuenta con una gran cantidad de librerías, tipos de datos y funciones incorporadas en el propio lenguaje, que ayudan a realizar muchas tareas habituales sin necesidad de tener que programarlas desde cero.



Usado por docentes, científicos e investigadores.

### Características del lenguaje

__(1) Lenguaje interpretado__: el ordenador es capaz de ejecutar la sucesión de instrucciones dadas por el programador
Para que esto funcione, necesitamos un __intérprete__, que es el programa encargado de traducir cada instrucción escrita en el lenguaje Python en instrucciones del sistema.

__(2) Multiplataforma__: Python es un lenguaje disponible en los principales Sistemas Operativos (Windows, Linux y Mac). 

Otras características importantes:

__(3) Multiparadigma__: Es un lenguaje imperativo y orientado a objetos. También posee características de los lenguajes funcionales.

__(4) Tipado dinámico__: El tipo de los datos es inferido en tiempo de ejecución. Esta es una característica que diferencia a Python de la mayoría de los lenguajes compilados.




En la siguiente Sección introducimos los tipos básicos del lenguaje Python, así como los conceptos de variable y asignación.

## Objetos, expresiones y tipos numéricos

Los __Objetos__ son las cosas que los programas escritos en Python pueden manipular. Todo objeto tiene asociado un __tipo__ (`type`), el cual define el tipo de operaciones que se pueden realizar con ese objeto.

Los tipos pueden ser de dos clases:

* __Objetos escalares__: Son objetos indivisibles. Se les considera los átomos del lenguaje. Un ejemplo son los números.
* __Objetos NO escalares__: Son objetos con una estructura interna. Por ejemplo las cadenas de caracteres, las listas, etc.

### Objetos escalares

Como tantos otros lenguajes de programación, Python ofrece, de entrada, los tipos de datos enteros, reales, booleanos y cadenas de caracteres. También permite el manejo de números complejos sin importar librería alguna. Veamos algunas operaciones básicas con estos tipos de datos.

#### 1. Objetos de tipo `int`.

El tipo `int` se usa para representar números enteros. Por ejemplo: `3`, `234`.

```python
# operaciones típicas entre enteros
(3+7, 3-7, 3 + 2*7, (3+2)*7, 1 + 2 * 3**2)
```

```python
# para conocer el tipo de un objeto
type(25)
```

####  2. Objetos de tipo `float`.

El tipo `float` se usa para representar números reales. Los números reales en Python incluyen la notación punto (`.`) para separar la parte entera de la parte decimal.

```python
43.897, 45.76 / 2
```

#### 3. Objetos de tipo `bool`.



El tipo `bool` se usa para representar las constantes booleanas `True` y `False`. 

```python
a = True  # cierto   = 1
b = False  # falso   = 0
```

#### 4. El tipo `None`

El tipo `None` representa un dato sin tipo. Ya hablaremos de él más adelante.

### Objetos no escalares

Ejemplos de objetos no escalares son: Las listas, tuplas, diccionarios, ect. También todos aquellos creados por el usuario. 

### Operaciones 

Los objetos y los __operadores__ se combinan para formar __expresiones__. Cada expresión es evaluada por el intérprete de Python para producir un resultado de un cierto tipo. Este resultado se conoce como __valor__ de la expresión. 



La siguiente tabla describe los operadores disponibles para los tipos de objetos `int` y `float`. Podemos ver su aridad (número de operandos), asociatividad y precedencia.


| Operador    | Operación        |  Precedencia  | Asociatividad   | Ejemplo     | Resultado    |
| :---------: | :----------------| :-----------: | :-------------  |:---------: | :----------: |
| +           | Suma             |  3            | Por la izquierda|5 + 2       | 7            |
| -           | Resta            |  3            | Por la izquierda|5 - 2       | 3            |
| *           | Multiplicación   | 2             | Por la izquierda|5 * 2       | 10           |
| /           | División         | 2             | Por la izquierda|5 / 2       | 2.5          |
| //          | División entera  | 2             | Por la izquierda|5 // 2      | 2            |
| %           | Módulo           | 2             | Por la izquierda|5 % 2       | 1            |
| **          | Potencia         | 1             | Por la derecha  | 5 ** 2      | 25           |

Los operadores de __suma__ (`+`), __resta__ (`-`), __multiplicación__ (`*`) y __división__ (`/`) funcionan del modo habitual. Los últimos 3 operadores no son tan comunes:

- La __división entera__ (`//`) devuelve el cociente de la división entera. Por ejemplo, 7 // 3 devuelve 2 porque 7 dividido por 3 tiene cociente 2 y resto 1.
- El __módulo__ (`%`) devuelve el resto de la división entera. Por ejemplo, 7 % 2 devuelve 1.
- La __potencia__ (`**`) eleva un número a otro. Por ejemplo, 5 ** 3 devuelve 125 porque 5 elevando al cubo es 125.


__Ejemplos:__

```python
# potencia 
2 ** 3      # 2 elevado a 3
```

```python
# resto de la división entera (operación módulo)
5 % 2
```

```python
# división entera: El resultado es un número sin decimales
5 // 2
```

```python
# división : El resultado es un número sin decimales
5 / 2
```

Los operadores booleanos son los habituales, `and` para la conjunción, `or` para la disyunción y  `not` para la negación.

![LPO](./images/lpo.png "LPO") 

__Ejemplos:__

```python
a = True  # cierto   = 1
b = False  # falso   = 0
# conjunción lógica (and) y disyuncion lógica (or):
a and b, a or b, a or True, b and False 
```

Python dispone de los operadores habituales de comparación. Los operadores de comparación son:

* `==` igual a
* `!=` distinto de 
* `<` menor que
* `<=` menor o igual que
* `>` mayor
* `>=` mayor o igual

Devolverán un valor booleano: `True` o `False`.

__Ejemplos:__

```python
34 != 34, not (34 == 34), a * 3 >= 7, b * 2 < 8, b * 2 <= 8, a >= 2 or b < 7
```

El operador `==` se utiliza para comprobar si el valor de dos expresiones son iguales.

## Variables y asignación

En ocasiones deseamos que el ordenador recuerde ciertos valores para usarlos más adelante. Necesitamos introducir el concepto de __variable__. Las variables proporcionan una forma de dar nombre a los objetos. Una variable se puede ver como  una etiqueta que está conectada con un `valor`.

```python
# Podríamos haber trabajado igualmente con variables:
dividendo = 17
divisor   = 5
cociente  = dividendo // divisor
resto    = dividendo % divisor

# Y sería bonito comprobar que el dividendo es igual a divisor * cociente + resto:

dividendo, divisor * cociente + resto
```

* Las dos primeras instrucciones conectan el nombre `dividendo` y `divisor` a un objeto de tipo `int`
* La tercera instrucción conecta el nombre `cociente` al valor de la expresión ` dividendo // divisor`, que es un objeto de tipo `int`.

Como hemos podido ver en el bloque de código superior, la forma de conectar un nombre con un valor es mediante el operador `=`, denominado operador de __asignación__. 



### Asignación múltiple

Python permite realizar la asignación múltiple. Por ejemplo, las dos celdas de código siguientes son equivalentes::

```python
x = 1
y = 2
z = 3
```

```python
x, y, z = 1, 2, 3
```

Las expresiones en el lado derecho se evaluan y se asocian a las variables en el lado izquierdo.

```python
print(y)
```

## Funciones predefinidas en Python

El siguiente bloque de código utiliza la función `print`:

## Cadenas de caracteres. El tipo `str`

Mucha gente usa Python por sus capacidades para el tratamiento y procesamiento de texto. 
Los mensajes de texto se denominan __cadenas de texto__ y representan secuencias de caracteres. Por ejemplo, la cadena de texto `Hola` representa una secuencia de 4 caracteres: `H`, `o` , `l ` y `a `. Las cadenas de texto se pueden escribir indistintamente entre comillas dobles (`"`) o comillas simples (`'`). 

Las cadenas de texto se corresponden con el tipo `str` en Python. Se trata de un tipo de objetos no escalar, con estructura interna. Los objetos de tipo `str` están compuestos por una secuencia de carateres.

El tipo cadena es [UNICODE](https://es.wikipedia.org/wiki/Unicode) por defecto.

```python
objetivo = "Aprender a manejar Python"
meta = 'Ser todo un experto'   # Las cadenas de caracteres se pueden poner con comillas dobles o simples.
print(objetivo)
print(meta)
```

Podemos escribir cadenas de texto que ocupen varias líneas usando comillas triples:

```python
print("""Hola,
me llamo Iñigo Montoya,
tu mataste a mi padre, 
prepárate a morir""")
```

Los objetos de tipo `str` se pueden comparar. El intérprete de Python utiliza el orden lexicográfico (siendo las mayúsculas 'anteriores' a las minúsculas:).

```python
a = 'Ana'
b = 'ana'
c = "Miguel"
a < b, a < c
```

Las cadenas de caracteres se pueden concatenar con el operador `+`. Es decir, el operador `+` devuelve otra cadena que es el resultado de concatenar las dos cadenas que lo rodean.

```python
a = 'Ana'
b = "Me he enamorado de " + a
b
```

El operador de repetición (`*`) construye cadenas repitiendo otra cadena varias veces:

```python
saludo = 'Hola ! ' * 3    # replicación
saludo
```

El tipo `str` es uno de los tipos secuencia que existen en Python. Otros tipos secuencia son : `list` para representar listas y  `tuple` para representar tuplas. Ambos tipos los veremos más adelante.

Los objetos de  tipo secuencia comparten una serie de operaciones.


| Operador    | Operación         | Ejemplo                | Resultado    |
| :---------: | :---------------- | :--------------------: | :----------: |
| +           | Concatenar        | 'Aprender ' + 'Python" | 'Aprender Python'   |
| *           | Repetir           | 2 * 'ja '      | 'ja ja '   |
| len()       | Longitud          | len('Hola')   | 4            |
| []          | Acceso a subcadena         | 'Hola'[1]       | 'o'          |
| [:]         | Acceso a subcadena         | 'vacaciones'[1:3]     | 'ac'         |

* `len`: Para conocer el tamaño de una cadena, se utiliza la función `len` como se muestra a continuación:

```python
a = 'Ana'
len(a)
```

* Indexación: Acceso a posiciones de una cadena de caracteres. Podemos acceder mediante indexación, es decir, usando la notación de corchetes `[ ]` e indicando la posición a la que queremos acceder. 

> Ojo, en Python la indexación empieza en 0.

```python
c = 'FRIDAY'
c[0]
```

El operador de corchetes `[ ] ` también se usa para acceder a un subcojunto de caracteres dentro de una cadena. Por ejemplo, si queremoa acceder a los tres primeros caracteres tenemos que escribir lo siguiente:

### Caracteres especiales

Dentro de las cadenas de texto podemos incluir algunos caracteres que se interpretan de manera especial al mprimirlos. Estos caracteres aparecen precedidos por el símbolo `\` :

### Formateando cadenas

El habitual cuando se trabaja con texto, tener la necesidad de construir una nueva cadena de caracteres a partir de otras más pequeñas. Una opción, como hemos visto anteriormente, es usar el operador `+`.

```python
texto = 'Importe total: '  + '100'
texto
```

Sin embargo, la forma moderna y recomendada de formatear cadenas de texto es mediante el uso de `f-strings` junto con los operadores de formato.

```python
radio = 4.323343
pi = 3.1416
area = pi*radio**2
texto = f'El área de un círculo de radio {radio} es {area}'
texto
```

También podemos usar modificadores de formato dentro de las llaves indicar un formato concreto, por ejemplo para ajustar los decimales:

```python
texto = f'El área de un círculo de radio {radio:.2f} es {area:.1f}'
texto
```

>  __Operadores de formato__

>  `:d` : ha de sustituirse por un número

>  `:s` : ha de sustituirse por un string

>  `:.2f` : ha de sustituirse por un número con 2 decimales

>  `:<` : alineación a la derecha

```python
importe1 = 15248.12
importe2 = 12.4545
texto1 = f'Importe mes de Enero:   {importe1:>10.2f}'
texto2 = f'Importe mes de Febrero: {importe2:>10.2f}'
print(texto1)
print(texto2)
```

## Errores comunes

__Ejemplo 1:__

```python
# El siguiente ejemplo te invita a reflexionar sobre la precisión
# de las operaciones y sobre la comparación de números reales
import math

def paradoja(a: float):
    b = math.sqrt(a)  # la raíz cuadrada de 2
    return a == b * b # esto debería ser cierto siempre, no lo es por el error de la representación de los reales

paradoja(2)
```

Es necesario usar la función `math.isclose` 

```python
def paradoja(a: float):
    b = math.sqrt(a)  # la raíz cuadrada de 2
    return math.isclose(a, b * b) 
paradoja(2)
```

__Ejemplo 2:__

```python
# Y otro más. Lo siguiente, matemáticamente,
# debería dar como resultado un entero:

def fibonacci(n):
    phi = (1 + math.sqrt(5)) / 2
    return (phi ** n - (1 - phi) ** n)/math.sqrt(5)

fibonacci(4)
```

```python
# Una solución sencilla consiste en redondear:

def fibonacci(n):
    phi = (1 + math.sqrt(5)) / 2
    return int(round((phi ** n - (1 - phi) ** n)/math.sqrt(5)))

fibonacci(4)
```
