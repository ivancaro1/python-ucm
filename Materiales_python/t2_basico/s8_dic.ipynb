# Diccionarios

---


Un **diccionario** (también llamado **memoria asociativa**) es parecido a una lista, pero en vez de dar información a partir de un índice, que es la posición, da información a partir de una clave cualquiera.
Las claves se comportan como elementos de un conjunto.
- No puede haber claves repetidas
- Las claves deben ser objetos con una función hash definida: `int`, `float`, `tuple`,... no listas.

```python
# Mis teléfonos:
    
telefonos = {"despacho": 3957, "casa": 917121314}
print(telefonos)
telefonos["movil"] = 687001122
print(telefonos)
print(telefonos["movil"])
del telefonos["movil"]
print(telefonos)
telefonos["casa"] = 91919111
print(telefonos)
```

```python
# Consulta de un elemento inexistente:

print(telefonos)
print(telefonos["manuel"])   # Esto dará lugar a un error
```

```python
# Se puede comprobar primero si está (la clave de) el elemento:

if "manuel" in telefonos:
    print(telefonos["manuel"])
else:
    print("manuel" + " no tiene un teléfono registrado")
```

## Recorridos en diccionarios

---

```python
for sitio in telefonos:   # es lo mismo que for sitio in telefonos.keys()
    print(sitio, telefonos[sitio])
```

```python
for num_tel in telefonos.values():
    print(num_tel)    
```

```python
for sitio, num_tel in telefonos.items():
    print(sitio, num_tel)    
```

```python
# A cada clave se le puede asignar una estructura de datos más compleja:
    
agenda = {"Fernando": {"tel.casa":  11111111, 
                       "tel.movil": 22222222, 
                       "direccion": "Pozuelo"
                      },
          "Elena":    {"tel.casa":  555555555, 
                       "tel.movil": 66666666, 
                       "direccion": "Colón"
                      }
         }
for persona in agenda:
    print(persona + " vive en " + agenda[persona]["direccion"])
```
