## Bucles  `while`

La instrucción `while` permite ejecutar un bloque entero de código tantas veces como queramos. Consta de dos cosas:

* Una condición de continuación (es decir, lo que tiene que ocurrir para que sigamos repitiendo.
* Un bloque de código que vamos a repetir.

Las condiciones funcionan exactamente igual que en las comparaciones con `if`: son una expresión booleana (algo que devuelve verdadero o falso). 
* Las instrucciones se repiten mientras se verifique la condición. 
* Si la expresión devuelve falso (`False` en Python), el bucle no volverá a repetirse y continuará con la instrucción que siga al `while`.

El número de repeticiones (también llamadas iteraciones) es variable y depende de la condición.


````python
# sumar los primeros números enteros hasta uno dado
lim_sup = 10
suma = 0
i = 1                 # inicialización 
while i <= lim_sup:   # condición
    suma = suma + i
    i = i + 1         # incremento

print(suma)
````

__Analizando la instrucción__:

* __Inicialización__: para llevar la cuenta de las veces que se repiten las intrucciones internas, creamos una variable `i` con un valor inicial a 1.

* __Condición__: Lo primero que hace el bucle `while` es comprobar si la condición (i <= lim_sup) es cierta. El cuerpo del bucle se ejecutará mientras la condición  `i <= lim_sup` , sea cierta.

* __Incremento__: Incrementamos el valor de `i` 

Como regla y para evitar bucles infinitos, debemos asegurarnos de que después de cada repetición, <span style='color: blue;'>la condición del `while` está más cerca de hacerse falsa </span>. En el ejemplo anterior, en el cuerpo del bucle se incrementa la variable `i`, por lo que cada vez el valor de `i` se aproxima más a *lim_sup*.

### Ejemplo 1

Sumar los números enteros en un determinado rango.

````python
def sumar(inf, sup):
    suma = 0      
    while inf <= sup:
        suma = suma + inf
        inf = inf + 1
    return suma    
````

Escribimos la función con sus comentarios de  ayuda:

````python
def sumar(inf:int, sup:int)->int:
    """
    This function calculates the sum of all integers from inf to sup,
    including both. If sup<inf this function returns 0 (there are no
    integers holding the conditions.
    Parameters
    ----------
    inf : Start of the range
    sup : Finish of the range
    Returns
    -------
    Sum of the range [a..b]
    Example
    -------
    >>> sumar(1,4)
    10
    """
    suma = 0      
    while inf <= sup:
        suma = suma + inf
        inf = inf + 1
    return suma    
````

### Ejemplo 2

Descomponer un número en dos partes:
 * Las potencias de 2
 * el resto

Por ejemplo: 
* el 6 se descompone en (2,3) ya que 6 = 2 * 3
* el 8 se descompone en (8,1) ya que 8 = 8 * 1
* el 24 se descompone en (8,3) ya que 24 = 8 * 3


````python
def potencias_de2(n):
    potencias = 1
    while (n % 2) == 0:
        potencias = potencias * 2
        n = n // 2
    return potencias, n
````

````python
potencias_de2(4)
````

````python
potencias_de2(1)
````

````python
potencias_de2(24)
````

### Ejemplo 3

Otra versión que hace algo un poco diferente. La  función debe devolver el valor $k$ tal que $n = 2^k * m$ .

__Ejemplos de ejecución__:

* `potencias_de2(4)`  devuelve  `2`
* `potencias_de2(8)`  devuelve  `3`
* `potencias_de2(24)`  devuelve  `3`

````python
def potencias_de2(n:int)-> int:
    """
    This function extract the greatest power of two of a given number n >= 0.
    It returns the exponent of the power of two, that is
    pow2(n) = s, k then (2**s)*k = n.
    Parameters
    ----------
    n: Integer to extract the greatest power of two
    Example
    -------
    >>> potencias_de2(12)
    2
    """
    cont = 0
    while (n % 2) == 0:
        cont = cont + 1
        n = n // 2
    return cont
````

````python
potencias_de2(4)
````

````python
potencias_de2(8)
````

````python
potencias_de2(24)
````

### Ejemplo 4

Suma de las cifras de un número. Con un poco de paciencia podemos sumar las cifras de un entero positivo utilizando únicamente expresiones y asignaciones. Con `while` la cosa es más sencilla, general y clara.

````python
def digit_sum(n:int)-> int:
    """
    This function computes the sum of the digits of a positive integer, n >= 0.
    Parameters
    ----------
    n : Positive integer to sum the digits
    Returns
    -------
    Sum of the digits of n
    Example
    -------
    in: digit_sum(123)
    out: 6
    """
    result = 0
    while n != 0:
        digit = n % 10
        result = result + digit
        n = n // 10
    return result
````

````python
digit_sum(152)
````

````python
digit_sum(0)
````

Un número es divisible por 3 si la suma de sus cifras es 0, 3, 6 ó 9.

````python
def div3(n):
    """
    This function decides if a positive integer is divisible by 3. n >= 0.
    Parameters
    ----------
    n : int
    Integer positive number
    Returns
    -------
    bool
    Whether n is divisible by 3 or not
    Example
    -------
    >>> divisible_by_3(14)
    False
    """
    while n > 9:
        n = digit_sum(n)
    if n == 0 or n == 3 or n== 6 or n == 9:
        return True
    else:
        return False
        
````
