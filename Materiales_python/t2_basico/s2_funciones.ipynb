# Funciones en Python

---

En Python la unidad de estructuración de los programas son las funciones. Una función consiste en un conjunto de instrucciones que realizan una determinada acción a las que se les asocia un nombre. Una vez definida la función, se puede usar (llamar o invocar) mediante su nombre. En general, la función devolverá como resultado un valor de retorno (aunque no siempre tiene que ser así).

Las funciones tienen un nombre y se declaran con la palabra reservada def y devuelve un valor usando la palabra reservada return.

También tienen una lista de `parámetros abstractos` (también llamados `parámetros formales`). Dichos parámetros pueden ser de varios tipos:

* posicionales
* por clave
* argumentos agrupados



## Funciones con parámetros posicionales

---

Definimos una función sencilla:

````python
# Definición de una función:
def formar_numero(dec:int, uni:int) -> int:
    valor_decenas = 10 * dec
    valor_unidades = uni
    return valor_decenas + valor_unidades

# Distintas llamadas a la función anterior:

formar_numero(3, 4), formar_numero(9, 2), formar_numero(0, 5)
````

* Las funciones tienen una primera línea que sirve para definirlas, indicando su nombre y los valores que la función necesita para realizar su tarea. Dichos valores, llamados `parámetros abstractos` se encuentran dentro de los paréntesis. La definición de la función siempre termina con dos puntos. Se declaran con la palabra reservada `def`. 
 
* Las funciones pueden devolver valores usando la palabra reservada `return`.

* Cada uso (o llamada) de la función se hace con los parámetros concretos con los que realmente necesitamos realizar los cálculos. En cada llamada, los parámetros abstractos (también llamados parámetros formales) asumen los valores de los `parámetros concretos` (también llamados `parámetros reales`) y se ejecuta el cuerpo con ellos.

Veamos algunos ejemplos de funciones:

````python
def media(x:float, y:float) -> float:
    return (x + y) / 2

media(3,25), media(3.4,5.6), media(-1,-3)
````

````python
a = 1
b = 10
 # los valores de x e  y se sustituyen por el ressultado de las
 # expresiones a * 2  y b // 2 respectivamente en la llamada
resultado = media (a * 2,  b // 2)  
resultado
````

Cuando se define una función, se crea una variable con el mismo nombre cuyo valor tiene tipo `function`. Así, si se introduce únicamente el nombre de una función sin paréntesis se imprime información de la función en vez de ejecutarse.


## Funciones con parámetros por clave

---

Los parámetros por clave se usan para indicar **valores por defecto** y siempre se sitúan **después de los parámetros posicionales**. De esta forma podemos crear una función que puede ser invocada con menos parámettros de los que permite.

````python
# función que suma 4 números y devuelve el resultado
def suma_varios(x, y, z1 = 0, z2 = 0):       # 2 parámetos posicionales y 2 por clave
    m = x + y + z1 + z2
    return m
````

````python
# Esta función puede ser invocada pasando sólo los argumentos obligatorios x e y 
resultado1 = suma_varios(2, 3)    # z1 y z2 tomarán los valores por defecto
resultado1
````

````python
# puede ser invocada pasando los argumentos obligatorios x e y junto con un argumento por clave
resultado2 = suma_varios(2, 3, z2 = 1)    # z1 toma el valor por defecto 0
resultado2
````

````python
# puede ser invocada pasando los argumentos obligatorios x e y junto con los dos argumento por clave
resultado3 = suma_varios(2, 3, z2 = 1, z1 = 9)   # ningún valor por defecto
resultado3
````

## Valores devueltos por una función

---

Las funciones pueden devolver múltiples valores. 

````python
def orden(a,b):
    if a <= b:
        return a, b  
    else:
        return b, a
````

````python
orden(7, 1)
````

Una función puede tener varias instrucciones `return`.  Si la ejecución de una función no alcanza ninguna instrucción `return`, se devuelve `None` que es el valor que representa el valor nulo en Python. El tipo de `None` es `NoneType`.

````python
def suma_todos(x, y, z, t):
    resultado = x + y + z + t     # calcula el resultado pero no devuelve nada
                                  # No hay instrucción return
````

````python
r = suma_todos(2, 3, 4, 6)
r                                    # r es None
````

````python
type(r)
````

La ejecución de una función termina en el momento en el cual se alcance una instrucción `return`.

````python
# función que pretende escribir cuatro cadenas: Uno, Dos, Tres, Cuatro
# si añadimos una instrucción return, no se ejecutan todas las instrucciones
def cuenta():
    print('Uno')
    print('Dos')
    return ('Fin de la función')  # la función termina tras ejecutar el return
    print('Tres')
    print('Cuatro')
````

````python
cuenta()
````

## Funciones con parámetros agrupados por tupla

---

Son parámetros posicionales agrupados que se representan mediante una cadena con un asterisco al principio de la misma (como por ejemplo `*args`). Aparece a continuación de los parámetros posicionales. Representa una tupla arbitraria de parámetros formales agrupados.

````python
def ejemplo_f(x: int, y:int , *args: tuple[int]):
    print( "x:", x )
    print( "y:", y )
    print("argumentos agrupados:", args)

ejemplo_f(1,2,3,4,5,6,7,8,9)
````

````python
def promedio(*args):
    return sum(args) / len(args)

promedio(7, 8, 9, 10, 6, 8)
````

## Funciones con parámetros agrupados nombrados

---

Son argumentos que se representan mediante una cadena con dos asteriscos al principio de la misma (como por ejemplo `**kwargs`). Aparece a continuación de los parámetros agrupados por tupla. Estos argumentos se almacenan en un diccionario donde la clave es el nombre del parámetro y el valor es el dato asociado.

````python
def ejemplo_g(x, y, *tupla, **kwargs):
    print( "x:", x )
    print( "y:", y )
    print("agrupados por tupla:", tupla)
    print("agrupados por diccionario:", kwargs)

ejemplo_g(1,2,3,4,5,6,7,8,9, cien = 100 , mil= 1000 )
````

````python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 15, 7, 20, 12]

# Gráfico de barras
def dibujar_grafico(x, y, **kwargs):
    tipo = kwargs.get("tipo", "linea")
    
    plt.figure(figsize=(5, 3))
    if tipo == "barras":        
        plt.bar(x, y, color=kwargs.get("color", "green"))
    elif tipo == "dispersion":
        plt.scatter(x, y, color=kwargs.get("color", "red"))
    else:
        plt.plot(x, y, color=kwargs.get("color", "blue"))
    
    plt.title(kwargs.get("titulo", "Gráfico"))
    plt.show()


dibujar_grafico(
    x, y,
    tipo="barras",
    color="orange",
    titulo="Ventas por mes"   
)
````

La llamada anterior crea internamente el diccionario:
```python
        kwargs = {
            "tipo": "barras",
            "color": "orange",
            "titulo"= "Ventas por mes"
        }
```

La función lee esos parámetros y decide qué gráfico usar (plot, bar o scatter). Esto es muy parecido a cómo funcionan muchas librerías profesionales, que aceptan decenas de opciones mediante **kwargs.
