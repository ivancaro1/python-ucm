## Condicionales: Instrucciones if, elif, else

La instrucción condicional __if__ se utiliza para comprobar una condición. 

* Si la condición es verdadera entonces procesaremos un bloque de instrucciones (llamado bloque if). 
* Si la condicicón no se cumple, se ejecuta el bloque de instrucciones alternativas (llamado bloque else). 
* Cada uno de los bloques están sangrados, sin necesidad de escribir llaves o sentencias del tipo `begin` ... `end`.

* La cláusula `else` es opcional.

Recordar que en Python los bloques se delimitan por sangrado. 
* Cuando ponemos los dos puntos al final de la primera línea del condicional, todo lo que vaya a continuación con __un__ nivel de sangrado superior se considera dentro del condicional. 
* En cuanto escribimos la primera línea con un nivel de sangrado inferior, hemos cerrado el condicional. 

__Ejemplo:__

````python
def aprobado(nota: float) -> bool:
    if nota >= 5.0:
        return True
    else:
        return False

aprobado(3.0), aprobado(7.5), aprobado(5.0), aprobado(10.0), aprobado(4.999) 
````

La función anterior está pobremente documentada, y queremos progresar tan rápido que a menudo omitimos la documentación. Hela aquí de nuevo con una documentación más completa:

````python
def aprobado(nota: float) -> bool:
    """
    Averigua si una nota numérica es suficiente para aprobar o no
    
    Parameters
    ----------
    nota : float
        La nota numérica
    
    Returns
    -------
    bool 
        True, si la nota es suficiente para aprobar
        False, si la nota no lo es
        
    Precondition
    ------------
    nota >= 0 and nota <= 10

    Examples
>>> aprobado(3.0)
False

>>> aprobado(7.5)
True

>>> aprobado(5.0)
True

>>> aprobado(4.999)
False
    """
    if nota >= 5.0:
        return True
    else:
        return False

aprobado(3.0), aprobado(7.5), aprobado(5.0), aprobado(10.0), aprobado(4.999) 
````

__Ejemplo:__

La rama `else` no siempre es necesaria:

````python
def valor_abs(x):
    """
    Returns the absolute value of x
    Parameters
    ----------
    x : int
    A number
    Returns
    -------
    int
    the absolute value of x
    """
    if x < 0:
      # cambio de signo del dato de entrada
      x = -x
    return x
````

````python
valor_abs(-5)
````

````python
valor_abs(8)
````

Otro ejemplo:

````python
def max_min(x, y):
    if x < y :
        max = y
        min = x
    else:
        max = x
        min = y
    return max, min

print(max_min(2, 3), max_min(3, 2))
````

__Cuidado con la indentación:__

````python
if 1 > 0:                            # esta condición siempre es falsa
    print("1 es menor que 0")
print("Esto se ejecuta siempre")  # esto no está dentro del bloque y no se ejecuta 
                                                    # dentro del if
````

__Elecciones múltiples:__

Si queremos añadir ramas adicionales al condicional, podemos emplear la cláusula `elif` (abreviatura de *else if*). Para la parte final, que debe ejecutarse si ninguna de las condiciones anteriores se ha cumplido, usamos la instrucción `else`. Veamos algunos ejemplos:

````python
def calificacion(num_nota: float) -> str:
    """
    Converierte una calificación numérica en su denominación
    
    Parameters
    ----------
    num_nota : float
        la nota numérica
    
    Returns
    -------
    str 
        la denominación de la calificación
        
    Precondition
    ------------
    num_nota >= 0 and num_nota <= 10

    Examples
    --------
>>> calificacion(10)
Sobresaliente

>>> calificacion(1.23)
Suspenso

>>> calificacion(3.45)
Suspenso

>>> calificacion(4.999)
Suspenso

>>> calificacion(7)
Notable

>>> calificacion(5)
Aprobado
    """
    if num_nota < 5:
        return "Suspenso"
    elif num_nota < 7:
        return "Aprobado"
    elif num_nota < 9:
        return "Notable"
    else:
        return "Sobresaliente"
    
calificacion(10), calificacion(1.23), calificacion(3.45), calificacion(4.999), calificacion(7), calificacion(5)
    
````

## La instrucción `pass`

La instrucción `pass` no hace nada. Se puede usar cuando la sintaxis obliga a escribir al menos una instrucción pero el programa no necesita ninguna acción.

Las sentencias `if` no pueden estar vacías; sin embargo, si por alguna razón tienes una sentencia `if` sin contenido, podemos incluir la sentencia `pass` para evitar errores.

````python
edad = 20

if edad < 18:
    pass  # TODO: Añadir más adelante la lógica para menores de edad
else:
    print("Acceso concedido")
````

La sentencia `pass` resulta útil en diversas situaciones:

* Cuando se está creando la estructura del código pero aún no se ha implementado la lógica
* Cuando una sentencia es sintácticamente necesaria pero no se requiere ninguna acción
* Como marcador de posición para código futuro durante el desarrollo
* En funciones o clases vacías que se planea implementar más adelante

## Expresiones ternarias

Las expresiones ternarias en Python tienen la siguiente forma:

```python 
    e = valorSiTrue if <condicion> else valorSiFalse
```

Permite definir la instrucción de la forma `if-else` en una sola línea. La expresión anterior es equivalente a:
```python
    if <condicion>:
        e = valorSiTrue
    else: 
        e = valorSiFalse
```

````python
a = 10
b = 20
bigger = a if a > b else b
print("Bigger is", bigger)
````

## Errores comunes
No cubrir todos los casos posibles.

````python
def val_absoluto_mal(x):
    if x < 0:
        return -x
    if x > 0:
        return x
    
val_absoluto_mal(0) == 0
````

Por tanto, es preferible, para las elecciones múltiples, usar "if .... elif ... elif ... else"
