# Uso de Ficheros

En nuestros programas, estamos acostumbrados a utilizar un gran número de datos, tanto de entrada como de salida. 

Es sencillo manejar los datos de manera interactiva, pero cuando tenemos una gran cantidad de datos, esto se puede volver incómodo. Podemos leer y almacenar los datos de manera permanente, utilizando [ficheros](https://docs.python.org/2/library/stdtypes.html#file-objects).



### Lectura de ficheros

Abrir el fichero [dataset.txt](./datos/dataset.txt) para leer su contenido. Usa la instrucción `open`. Usa el método `readlines()` para leer todas las lineas del fichero.
Muestra las lineas leídas.

```python
file = open('./datos/dataset.txt')
lines = file.readlines()
lines
```

```python
file = open('./datos/dataset.txt')
line = file.readline()
print(line, '===') #vemos el salto de línea
```

```python
file.close() ## Cuando se deja de usar un fichero, hay que cerrarlo
```

Con `readline()` leemos el fichero línea a línea.

```python
f = open('datos/maneras.txt', 'r')
linea = f.readline()
i = 1
while linea != '':
    print(i, linea)
    linea = f.readline()
    i += 1
f.close()
```

De hecho, el fichero se comporta un poco como una lista.

```python
f = open('datos/maneras.txt', 'r')
for linea in f:
    print(linea.strip()) #Eliminamos el carácter fin de línea   

f.close()
```

El fichero puede contener cualquier dato, por ejemplo números, pero tenemos que tener claro que, para el objeto fichero de python, todos los datos serán tratados como **cadenas de caracteres**.

### Excritura

También podemos guardar datos en un fichero. Al abrir un fichero para escribir pueden ocurrir dos cosas:
+    Si el fichero existía pierde su contenido anterior
+    Si el fichero no existía se crea con el contenido que le demos

```python
f = open('datos/mi_texto.txt', 'a')
f.write('Con cien cañones por banda')
f.write('Viento en popa a toda vela...')
f.close()
```

Por supuesto podemos crear un fichero para guardar números... pero como cadenas!

```python
l = range(50)
h = open('datos/lista_numeros.txt','w')
for x in l:
    h.write(f'{x}\n')

h.close()
```

## Evitando errores, instrucción with

La instrucción **with** evita errores de programación.
+  Hace el `close` del fichero de forma automática
+  Si se produce cualquier error, se llama a la instrucción `close`

```python
with open('datos/maneras.txt', 'r') as f:
    for linea in f:
        print(linea.strip())
```

```python
with open('datos/mi_texto2','w') as f:
    f.write('Con cien cañones por banda\n')
    f.write('Viento en popa a toda vela...\n')
```

Si se produce un error dentro del cuerpo del `with`, el fichero se cierra.

```python
lst = [3, 4, 5, 0, 6, 8]
with open('datos/inverses.txt', 'w') as f:
    for number in lst:
        f.write(f'{1/number}\n')
```

Ejemplo:

```python
from typing import Dict
import string

def frequency(filename: str) -> Dict[str, float]:
    def get_words(line):
        def clean_word(word):
            for c in string.punctuation + string.whitespace + '¿¡«»':            
                word = word.replace(c, '')
            return word.lower()
            
        return [clean_word(w) for w in line.split()]
        
    freq = dict()
    totalwords = 0
    with open(filename, encoding="utf-8") as fl:
        for line in fl:
            words = get_words(line)
            totalwords += len(words)
            for w in words:
                if w in freq:
                    freq[w] += 1
                else:
                    freq[w] = 1
    for word in freq:
        freq[word] = freq[word]/totalwords
    return freq
```

Puede producirse error en Windows porque Python intenta abrir un archivo utilizando la codificación por defecto del sistema (generalmente cp1252 o charmap) en lugar de UTF-8, que es el estándar en el que está guardado tu archivo. La solución pasa por añadir el parámetro `encoding="utf-8"` al momento de abrir el archivo.

```python
f = sorted(frequency('datos/quijote.txt').items(), key = lambda x: x[1], reverse = True)
print(f'El quijote tiene {len(f)} palabras')
print(f'Las palabras más fecuentes son:\n {f[:20]}')
print(f'Las palabras menos fecuentes son:\n {f[-20:]}')
```

## Ficheros zip

Para manejar ficheros zip, en python tenemos el módulo [`zipfile`](https://docs.python.org/3/library/zipfile.html) que es un módulo estándar. 
El primer problema es que tenemos el fichero en una secuencia de bytes. Para ello tenemos 2 clases del módulo [`io`](https://docs.python.org/3/library/io.html), 
* `StringIO`, que maneja cadenas de caracteres como si fueran ficeros de texto
* `BytesIO`, que manea secuencias de bytes. Es la necesitamos en este caso

```python
import zipfile, io
url = 'https://ckan0.cf.opendata.inter.prod-toronto.ca/dataset/7dda2235-999e-4a17-b228-abd0961e045d/resource/02c90a3a-d754-4023-a283-ed5687e87f1f/download/Traffic%20Signal%20Timing.zip'
r = requests.get(url)
```

```python
r.status_code, type(r.content), len(r.content)
```

```python
import zipfile, io

bytes = io.BytesIO(r.content)
zfile = zipfile.ZipFile(bytes)
```

```python
zfile.filelist
```

De `zfile` necesitamos extraer el contenido de `traffic-signals-timing.csv`

```python
f = zfile.open('traffic-signals-timing.csv')
lines = f.readlines()
f.close()
lines[:10]
```

La `b` delante de la cadena de caracteres indica que es una secuencia de bytes, no de str. Para trasformarla a cadena de caracteres con la función `decode`

```python
with zfile.open('traffic-signals-timing.csv') as f:
    lines = [ l.decode('utf-8') for l in f.readlines() ]
lines[:10]
```

Podemos leer todo el fichero, transformarlo en una cadena de caracteres y luego usar el método `StringIO`.

```python
import pandas as pd
with zfile.open('traffic-signals-timing.csv') as f:
    contents = f.read()
    contentstr = contents.decode('utf-8') # Decodifica los bytes a una cadena de texto (string) de Python utilizando el formato utf-8 para interpretar correctamente caracteres especiales y tildes.
    fstr = io.StringIO(contentstr)
    reader = pd.read_csv(fstr, delimiter = ',' )
    
reader
```
