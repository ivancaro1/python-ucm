## Ocultación de la información

Un aspecto importante en la programación orientada a objetos es la ocultación de la información.

El concepto de miembro privado (atributo o  método), que no puede accederse excepto desde dentro de un objeto, no existe en Python. Sin embargo, hay una convención que se sigue en la mayoría del código Python: 

* Un nombre prefijado con un guión bajo (por ejemplo, `_color`) debería tratarse como una parte no pública de la API. Debería considerarse un detalle de implementación y que está sujeto a cambios sin aviso. Se trata de una __protección suave__ (convención). En otras palabras, no es realmente privado, sino protegido y es más una advertencia para otros desarrolladores.
 
* Para una protección más estricta, se usa el doble guión bajo `__` (por ejemplo, `__matricula`).  Cuando el nombre de un atributo o método comienza con `__` pero no termina con `__`, ese atributo no es accesible desde fuera de la clase.


#### Atributos privados

```python
class Vehiculo:
    def __init__(self, precio:float, matricula:str):
        self._precio = precio           # atributo privado (suave)
        self.__matricula = matricula    # atributo privado (estricto)  

      
    def __str__(self):
        return f"Precio: {self._precio},\nMatrícula: {self.__matricula}" 
```

```python
coche = Vehiculo( 45,'ABC 0000')
print(coche)
```

Para permitir que el acceso a atributos privados mediante métodos de instancia, usamos el decorador `@property` . 

#### El decorador @property

* El decorador `@property` se utiliza para definir métodos de instancia que pueden ser accedidos como si fueran atributos de una clase, sin la necesidad de usar paréntesis. 
* El decorador `@property` convierte un método en una propiedad que puede ser accedida como un atributo.
* El uso del decorador `@property` convierte un método en un "getter" (para obtener el valor de una propiedad)

```python
class Vehiculo:
    def __init__(self, precio:float, matricula:str):
        self._precio = precio           # atributo privado (suave)
        self.__matricula = matricula    # atributo privado (estricto)  

    @property
    def precio(self)-> str:    # getter : devuleve el valor del atributo
        return self._precio 
    @property
    def matricula(self)-> str:   # getter : devuleve el valor del atributo
        return self.__matricula
        
    def __str__(self):
        return f"Marca: {self.precio},\nMatrícula: {self.matricula}" 
```

```python
coche = Vehiculo( 45,'ABC 0000')
print(coche)
```

```python
coche.matricula
```

#### Decorador `@nombre.setter`

El método decorado con `@nombre.setter` se usa para asignar un nuevo valor al atributo nombre. Este método permite agregar lógica de validación o procesamiento antes de modificar el atributo.

El uso de un `setter` con `@property` es útil porque te permite controlar el acceso y la modificación de los atributos de una clase sin exponerlos directamente. También mejora la encapsulación y permite agregar lógica adicional (como validación) al establecer los valores.

```python
class Vehiculo:
    def __init__(self, precio:float, matricula:str):
        self._precio = precio           # atributo privado (suave)
        self.__matricula = matricula    # atributo privado (estricto)  

    # getters
    @property
    def precio(self)-> str:
        return self._precio 
        
    @property
    def matricula(self)-> str:
        return self.__matricula
        
    # setters    
    @precio.setter
    def precio(self, valor)-> None:   # Actualiza el precio del coche
        if valor > 0:
            self._precio = valor
        else:
            pass
            
    @matricula.setter
    def matricula(self, valor)-> None:    # Actualiza matriculo
        m = valor.split(' ')
        if len(m[0]) == 3 and len(m[1]) == 4:        
            self.__matricula = valor
        else:
            self.__matricula = '000 0000'
        
    def __str__(self):
        return f"Precio: {self._precio},\nMatrícula: {self.matricula}" 
```

```python
coche = Vehiculo( 45,'ABC 0000')
print(coche)
```

```python
coche.matricula = 'ZZZ 1234'  
print(coche)
```

```python
coche.matricula = 'ZZ 1234'
print(coche)
```

## Métodos privados

```python
class Vehiculo:
    def __init__(self, marca, año, precio, matricula):
        self.marca = marca
        self.año = año
        self._precio = precio
        self.__matricula = matricula    # atributo privado

    def __str__(self):
        return f"Marca: {self.marca},\nMatrícula: {self.__matricula}"
           
    def __control_precio(self):
        return self.precio if self.precio != 0  else None    
    
```

```python
coche = Vehiculo('Tesla', 2022, 45,'ABC 0000')
coche.__control_precio()
```

## Efecto de la ocultación en las subclases

Observe que cuando una subclase intenta utilizar un atributo oculto de su superclase, se produce un error.


```python
class Vehiculo:
    def __init__(self, marca, año, precio, matricula):
        self.marca = marca
        self.año = año
        self.precio = precio
        self.__matricula = matricula    # atributo privado

    def __str__(self):
        return f"Marca: {self.marca},\nMatrícula: {self.__matricula}"
           
    def __control_precio(self):
        return self.precio if self.precio != 0  else None    
    
```

```python
class Coche(Vehiculo):
    def __init__(self, marca, año, precio, matricula, carroceria):
        super().__init__(marca, año, precio, matricula)
        self.carroceria = carroceria

    def __str__(self):
        info_vehiculo = super().__str__()
        return f"{info_vehiculo}, \nTipo de Carrocería: {self.carroceria}"
    
    def cambio_matricula(self, m):
        self.__matricula = m
        
    def validar_precio(self):
        self.__control_precio()
```

```python
coche_1 = Coche('Tesla', 2022, 45, 'ABB 3333','Sedán')
coche_1.cambio_matricula('ZZZ 1234')
print(coche_1)
```

```python
coche_1 = Coche('Tesla', 2022, 45, 'ABB 3333','Sedán')
coche_1.validar_precio()
```
