# Anotaciones de tipos

---

Las anotaciones de tipos (type hints) permiten declarar de forma explícita qué tipo de datos esperan los métodos y qué tipo de datos devuelven, así como los tipos de los atributos de una clase.

```python
class Usuario:
    # Atributo de Clase / Estático
    contador_usuarios: int = 0  

    def __init__(self, nombre: str, edad: int, activo: bool = True) -> None:
        # Atributos de Instancia
        self.nombre: str = nombre
        self.edad: int = edad
        self.activo: bool = activo
        
        # Atributo privado con tipo
        self._email: str | None = None
```

El método `__init__` siempre debe anotarse con retorno `None`.

Cada tipo de método tiene convenciones específicas para la anotación del parámetro que referencia a la instancia (self) o a la clase (cls):

```python
from typing import Self

class Producto:
    impuesto: float = 0.21

    def __init__(self, nombre: str, precio_base: float) -> None:
        self.nombre: str = nombre
        self.precio_base: float = precio_base

    # Método de Instancia: 'self' no se anota, pero los demás parámetros y el retorno sí
    def calcular_precio_total(self, descuento: float = 0.0) -> float:
        precio_con_descuento = self.precio_base * (1 - descuento)
        return precio_con_descuento * (1 + self.impuesto)

    # Método de Clase: 'cls' no se anota. Devuelve una instancia de la propia clase (Self)
    @classmethod
    def crear_gratis(cls, nombre: str) -> Self:
        return cls(nombre=nombre, precio_base=0.0)

    # Método Estático: Funciona como una función normal, todos sus parámetros se anotan
    @staticmethod
    def es_precio_valido(precio: float) -> bool:
        return precio >= 0.0
```
