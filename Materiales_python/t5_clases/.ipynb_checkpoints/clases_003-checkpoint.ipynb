## Tipos de métodos

En python existen varios tipos de métodos. 

### Métodos de instancia

Los `métodos de instancia` son métodos que reciben como parámetro la palabra `self`, que representa un objeto (o una instancia). Estos métodos solo pueden ser llamados a partir de un objeto de la clase.

Ejemplos de  métodos de instancia son los métodos llamados `especiales`, como son `__init__`, `__eq__`, `__str__`, etc. También son métodos de instancia todas las funciones que reciben `self` como parámetro.

```python
class Vehiculo:

    def __init__(self, c ='Blanco', a = 2.5, v = 0, p = 4):
        self._color = c
        self._ancho = a
        self._velocidad = v
        self._plazas = p

    def __str__(self):                      # método de instancia
        return f'Color {self._color}, ancho {self._ancho}' 
    
    def cambiarColor(self, nuevo_col):      # método de instancia
         self._color = nuevo_col

    def parar(self):                        # método de instancia
        self._velocidad = 0
```

### Métodos estáticos

Los métodos estáticos tienen la característica de que __no necesitan estar vincuados a un objeto__ de la clase para poder ser llamados. 

* Por tanto, estos métodos se corresponden con funciones que no reciben el  parámetro `self` en la definición. No dependen de los atributos de la clase.

* Se invocan con el nombre de la clase.

Estos métodos son marcados con un decorador llamado `@estaticmethod`. Este decorador se escribe justo antes de la definición del método estático y sirve para indicar que es un método estático.

```python
class Vehiculo:

    def __init__(self, c ='Blanco', a = 2.5, v = 0, p = 4):
        self._color = c if Vehiculo.validar_color(c) else 'Blanco'
        self._ancho = a
        self._velocidad = v
        self._plazas = p
    
    def __str__(self):    # método de instancia
        return f'Color {self._color}, ancho {self._ancho}' 

    def __repr__(self):   # método de instancia
        return f'Vehiculo({self._color}, {self._ancho},  {self._velocidad}, {self._plazas})'

    def parar(self):     # método de instancia
        self._velocidad = 0
    
    @staticmethod
    def validar_color(n)-> bool:
        return n in ['Rojo', 'Blanco', 'Azul']
 
```

```python
coche_1 = Vehiculo('Morado')
print(coche_1)
```

```python
coche_2 = Vehiculo('Azul')
print(coche_2)
```

Los métodos estáticos no están vinculados a un objeto de la clase, pero sí __están vincuados a la clase__, por lo que se llaman anteponiendo el nombre de la clase. 

Pueden ser llamados desde fuera de la clase o desde un método de instancia.



```python
Vehiculo.validar_color('Rojo')
```

```python
Vehiculo.validar_color('morado')
```
