# Ejercicios de Clases. 

## Ejercicio 1. 


Queremos crear un nuevo tipo de datos que represente una __cola de temperaturas__. Cada elemento de la cola es del  tipo `(sensor, temperatura)`, donde sensor es de tipo str y representa el nombre de un sensor y temperatura es la temperatura registrada por dicho sensor. Recordar que el comportamiento de la inserción y eliminación de registros en una cola implica que  los elementos se añaden por el final y la eliminación de elementos se hace por el principio (como una cola del supermercado). Por tanto, cuando un sensor `s` recoge una temperatura `t`, se añade a la cola como último elemento el registro `(s, t)`. Las temperaturas se van procesando en orden de llegada y una vez procesadas se eliminan de la cola.

Para implementar este nuevo tipo de datos usaremos la definición de clases y el mecanismo de herencia. 

* Crear una clase llamada `ColaTemp` que  herede de la clase base `list` de Python. Una cola es conceptualmente una estructura de datos que puede implementarse fácilmente sobre una lista. En Python, la clase list ya proporciona gran parte de la funcionalidad necesaria para manejar estructuras lineales, como agregar y eliminar elementos.
* Atributos:
    * num : número de elementos de la cola (número de registros de temperatura sin procesar.
* Métodos: 
    * Operación `len` para número de registros de temperatura sin procesar.
    * La operación `print` de una cola debe permitir que el usuario pueda ver el contenido completo de la cola y el número de elementos.
    * Registro de un nuevo elemento (inserción de un par  `(s, t)` al final.
    * Procesar un registro (eliminar el primer elemento de la cola).
    
* Métodos que no se permite invocar
    * No se permite procesar elementos intermedios (insert)
    * No se puede indexar o hacer slicing (cola[0]). Para esto es necesario imposibilitar la llamada a \__getitem\__ y \__setitem\__
    * No se puede modificar el atributo num desde fuera de la clase, pero si se puede consultar


* Comprueba qué pasa on la operación `in `.
