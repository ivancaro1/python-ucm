# Anotaciones de tipos

---

Las anotaciones de tipos (type hints) permiten declarar de forma explícita qué tipo de datos esperan los métodos y qué tipo de datos devuelven, así como los tipos de los atributos de una clase.

```python
class Usuario:
    # Atributo de Clase / Estático
    contador_usuarios: int = 0  

    def __init__(self, nombre: str, edad: int, activo: bool = True) -> None:
        # Atributos de Instancia
        self.nombre: str = nombre
        self.edad: int = edad
        self.activo: bool = activo
        
        # Atributo privado con tipo
        self._email: str | None = None
```

El método `__init__` siempre debe anotarse con retorno `None`.
