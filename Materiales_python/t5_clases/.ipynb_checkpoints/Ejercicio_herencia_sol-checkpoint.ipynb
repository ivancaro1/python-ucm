## Ejercicio 1. 


Queremos crear un nuevo tipo de datos que represente una __cola de temperaturas__. Cada elemento de la cola es del  tipo `(sensor, temperatura)`, donde sensor es de tipo str y representa el nombre de un sensor y temperatura es la temperatura registrada por dicho sensor. Recordar que el comportamiento de la inserción y eliminación de registros en una cola implica que  los elementos se añaden por el final y la eliminación de elementos se hace por el principio (como una cola del supermercado). Por tanto, cuando un sensor `s` recoge una temperatura `t`, se añade a la cola como último elemento el registro `(s, t)`. Las temperaturas se van procesando en orden de llegada y una vez procesadas se eliminan de la cola.

Para implementar este nuevo tipo de datos usaremos la definición de clases y el mecanismo de herencia. 

* Crear una clase llamada `ColaTemp` que  herede de la clase base `list` de Python. Una cola es conceptualmente una estructura de datos que puede implementarse fácilmente sobre una lista. En Python, la clase list ya proporciona gran parte de la funcionalidad necesaria para manejar estructuras lineales, como agregar y eliminar elementos.
* Atributos:
    * num : número de elementos de la cola (número de registros de temperatura sin procesar.
* Métodos: 
    * Operación `len` para número de registros de temperatura sin procesar.
    * La operación `print` de una cola debe permitir que el usuario pueda ver el contenido completo de la cola y el número de elementos.
    * Registro de un nuevo elemento (inserción de un par  `(s, t)` al final.
    * Procesar un registro (eliminar el primer elemento de la cola).
    
* Métodos que no se permite invocar
    * No se permite procesar elementos intermedios (insert)
    * No se puede indexar o hacer slicing (cola[0]). Para esto es necesario imposibilitar la llamada a \__getitem\__ y \__setitem\__
    * No se puede modificar el atributo num desde fuera de la clase, pero si se puede consultar


* Comprueba qué pasa on la operación `in `.



```python
class ColaTemp(list):
    """
    Implementación de la clase Cola de temperaturas de sensores
    """
    def __init__(self):
        self._num = 0
        # Inicializar la cola vacía como una lista vacía que almacenará puntos (x, y)
        super().__init__()
        
    # Sobrescribir el método __str__ para mostrar los registros y el número de registros
    def __str__(self):
        return f'{super().__str__()} \nNúmero de registros en cola: {self._num}'
        
    def __len__(self):
        return self._num

    # Método para insertar un registro de un sensor al final de la cola
    # Sobreescibimos el método append
    def append(self, sensor: str, temp: float) -> None:
        reg = (sensor, temp)
        super().append(reg)
        self._num += 1
        

```

```python
class ColaTemp(list):
    """
    Implementación de la clase Cola de temperaturas de sensores
    """
    def __init__(self):
        self._num = 0
        # Inicializar la cola vacía como una lista vacía que almacenará puntos (x, y)
        super().__init__()
        
    # Sobrescribir el método __str__ para mostrar los registros y el número de registros
    def __str__(self):
        return f'{super().__str__()} \nNúmero de registros en cola: {self._num}'

    def __len__(self):
        return self._num

    # Método para insertar un registro de un sensor al final de la cola
    def append(self, sensor: str, temp: float) -> None:
        reg = (sensor, temp)
        super().append(reg)
        self._num += 1

    # Método para eliminar el primer punto (primer elemento de la cola)
    def pop(self):
        if self._num > 0:
            registro = super().pop(0)
            self._num -= 1
            print(f"I - Registro {registro} procesado.")
            return registro
        else:
            return None
```

```python
r = ColaTemp()
len(r)
r.append('AS', 53.2)
r.append('B5', 1.5)
print(r)
r.pop()
print(r)
```

```python
from  typing import Union

class ColaTemp(list):
    """
    Implementación de la clase Cola de temperaturas de sensores
    """
    def __init__(self) -> None:
        self._num = 0
        # Inicializar la cola vacía como una lista vacía que almacenará puntos (x, y)
        super().__init__()
    
    @property
    def num(self):
        return self._num
    
    @num.setter 
    def num(self, valor:int) -> None:
        # No dejamos que se actualice desde fuera
        # solo los métodos defnidos en la clase pueden actualizarlo
        return None
        
    # Sobrescribir el método __str__ para mostrar los registros y el número de registros
    def __str__(self) -> str:
        return f'{super().__str__()} \nNúmero de registros en cola: {self._num}'

    def __len__(self)-> int:
        return self._num

    # Método para insertar un registro de un sensor al final de la cola
    def append(self, sensor: str, temp: float) -> None:
        reg = (sensor, temp)
        super().append(reg)
        self._num += 1

    # Método para eliminar el primer punto (primer elemento de la cola)
    def pop(self)-> Union[tuple[str, float], None]:
        """
        Examples
        --------
        >>> r = ColaTemp()
        >>> r.append('AS', 53.2)
        >>> r.pop()
        I - Registro ('AS', 53.2) procesado.
        ('AS', 53.2)
        >>> r.pop()
        """
        if self._num > 0:
            registro = super().pop(0)
            self._num -= 1
            print(f"I - Registro {registro} procesado.")
            return registro
        else:
            return None


    def insert(self, *arg) -> None:
        raise AttributeError( f'ColaTemp object has no attribute insert')

    def __getitem__(self, i)-> None:        
        # Este método permite acceder a un elemento por posición.
        # Aunque el enunciaco dice que no lo permitamos, lo dejo aquí 
        # para que se vea lo que habría que hacer
        """        
        Examples
        --------
        >>> r = ColaTemp()
        >>> r.append('AS', 53.2)
        >>> r[0]
        ('AS', 53.2)
        >>> r[5]
        Traceback (most recent call last):
          ...
        IndexError: list index out of range
        """
        return super().__getitem__(i)
        
        
    def __setitem__(self, *arg)-> None:
        raise AttributeError( f'ColaTemp object is not iterable')
        
import doctest
doctest.testmod(verbose=False)
```
