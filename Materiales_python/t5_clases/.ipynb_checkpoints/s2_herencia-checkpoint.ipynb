# Herencia

Muchos tipos tienen propiedades en común con otros tipos. Por ejemplo, el tipo lista y el tipo str tienen en común la función `len` que significan lo mismo. 

El mecanismo de herencia permite a los programadores definir una estructura jerárquica de clases, en la cual una clase hereda atributos y métodos de la clase superior en la jerarquía. No nos olvidemos que en Python la clase `Object` está en lo alto de la jerarquía. Todas las clases heredan de `Object`.

La herencia es un proceso mediante el cual se puede crear una clase hija que hereda de una clase padre, compartiendo sus métodos y atributos. Además de ello, una clase hija puede sobreescribir los métodos o atributos, o incluso definir unos nuevos.

Es un mecanismo exclusivo y fundamental de la POO. Es el principal mecanismo que ayuda a fomentar y facilitar la reutilización de código.

## La clase NIF

```python

import re
class Nif(str):
    def __init__(self, nif):
        
        Nif.validar(nif)      # validar el parámetro de entrada
        str.__init__(nif)     # llamada a init de la clase str


    @staticmethod
    def validar(dni):
        exp = "[0-9]{8}[A-Z]"
        DIGITO_CONTROL = "TRWAGMYFPDXBNJZSQVHLCKE"
        valido =   re.match(exp, dni) is not None and dni[8] == DIGITO_CONTROL[int(dni[0:8]) % 23] 
        if valido:
            pass
        else:
            raise ValueError(f'Número  {dni} no es un dni válido')
```

```python
nif_1 = Nif('72849506L')
print(nif_1)
```

```python
nif_2 = Nif('A28495066')
```

## Jerarquía de clases

En Python se puede crear una clase hija con tan solo pasar como parámetro la clase de la que queremos heredar. En el siguiente ejemplo vemos cómo se puede usar la herencia en Python, con las clases Coche, Camión y Motodicleta que heredan de Vehículo. 

```python
class Vehiculo:
    def __init__(self, marca, año, precio):
        self.marca = marca
        self.año = año
        self.precio = precio

    def __str__(self):
        return f"Marca:{self.marca}, Año:{self.año}, Precio:{self.precio:.2f}"
    
    def aplicar_descuento(self, d):
        self.precio -= self.precio * d / 100
```

```python
class Coche(Vehiculo):
    def __init__(self, marca, año, precio, carroceria, plazas):
        super().__init__(marca, año, precio)
        self.carroceria = carroceria
        self.plazas = plazas

    def __str__(self):
        info_vehiculo = super().__str__()
        return f"Coche-{info_vehiculo}, Tipo de Carrocería:{self.carroceria}, Plazas:{self.plazas}"
    
    def __len__(self):
        return self.plazas
```

```python
coche_1 = Coche('Tesla', 2022, 45, 'Sedán', 4)
print(coche_1)
```

```python
coche_1.aplicar_descuento(10)
print(coche_1)
```

### Método init en clase coche

El método `__init__` definido en la clase `Coche` hace dos cosas:
* invoca al método init de la clase padre `Vehículo` para inicializar los atributos heredados. Esto lo hacemos usando `super()`.  

`super()` es una función incorporada en Python que se utiliza para acceder y llamar a métodos y atributos de una clase padre (superclase) desde una subclase. Se utiliza comúnmente en el contexto de la herencia cuando tienes una clase que hereda de otra y deseas aprovechar el comportamiento de la superclase en la subclase.

* Después inicializa los atributos propios de los objetos de la clase coche que no están disponibles en la clase vehículo: tipo de carrocería y número de plazas.

### Método especial str

El método especial __str__ se define en la clase hija. 
En este caso, se invoca al método __str__ de la clase padre usando nuevamente super(). Posteriormente, se completa la información del objeto añadiendo tipo de carrocería y número de plazas.

### Método __len__

El método especial __len__ se define únicamente en la clase hija. 
Esto significa que la operación tendrá éxito para objetos de tipo coche y saltará una excepción para objetos de tipo Vehículo.

```python
len(coche_1)
```

### Método aplicar_descuento

Una clase hija que hereda de una clase padre, compartiendo sus métodos y atributos.
La clase `Coche` hereda de la clase padre los atributos y métodos. El método aplicar_descuento, definido en la clase Vehículo es heredado por coche.

Cuando se intenta evaluar el resultado de aplicar descuentto sobre el objeto coche_1, el intérprete busca en primer lugar el método en la clase 'Coche'. Si no lo encuentra definido allí, lo busca hacia arriba en jerarquía de clases. Dicho método se encuentra en la superclase, en la clase `Vehículo` y por tanto es éste el que se ejecuta.

```python
coche_1.aplicar_descuento(10)
print(coche_1)
```

## Añadir más clases a la jerarquía

```python
class CocheElectrico(Coche):
    def __init__(self, marca, año, precio, carroceria, plazas, bateria):
        super().__init__(marca, año, precio, carroceria, plazas)
        self.capacidad_bateria = bateria

    def __str__(self):
        info_coche = super().__str__()
        return f"{info_coche}, Capacidad de Batería: {self.capacidad_bateria} kWh"
    
    def cambiar_bateria(self, nueva_bateria):
        self.bateria = nueva_bateria
```

```python
class CocheGasoil(Coche):
    def __init__(self, marca, año, precio, carroceria, plazas, consumo):
        super().__init__(marca, año, precio, carroceria, plazas)
        self.consumo_combustible = consumo

    def __str__(self):
        info_coche = super().__str__()
        return f"{info_coche}, Consumo de Combustible: {self.consumo_combustible} litros/100 km"
```

```python
# Ejemplo de uso:

coche_electrico1 = CocheElectrico("Tesla", "Model S", 2022, 80000.0, "Sedán", 100)
coche_gasoil1 = CocheGasoil("Volkswagen", "Golf", 2022, 30000.0, "Compacto", 5.5)

print(coche_electrico1)
print(coche_gasoil1)
```
