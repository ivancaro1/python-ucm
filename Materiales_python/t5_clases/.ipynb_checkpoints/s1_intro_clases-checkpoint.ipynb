# Programación orientada a objetos

## Un poco de terminología

En el contexto de la programación, las `clases` son un mecanismo diseñado para construir tipos de datos similares a los objetos de la vida real. Si nuestros programas usan clases, hablamos de __programación orientada a objetos__.

Los lenguajes de programación orientados a objetos (Python lo es), permiten definir nuevos tipos de datos cuando los tipos proporcionados por el lenguaje no cubren todas las necesidades de la aplicación que estamos diseñando.

Recordar que los tipos de datos que hemos visto hasta ahora son:

* Objetos de tipo escalar: `int`, `float`, `bool`, `None`.
* Objetos de tipo no escalar: `list`,  `tuple`, `str`, `dict`.


Los __objetos__ son un mecanismo de los lenguajes de programación para unir datos y  funcionalidad: 

__Datos__:

* Un determinado objeto tiene un conjunto de características (denominadas __atributos__), cada una de las cuales tiene un valor. Estas características indican cuál es su estado actual.  
* Los objetos pueden tener una cantidad arbitraria de atributos de cualquier tipo.

__Funcionalidad:__

* Los objetos responden a un conjunto de operaciones (funciones ó __métodos__).

En Python, el núcleo de cosas que pueden manipular los programas son los objetos. Cada objeto tiene un tipo donde se definen las operaciones que se pueden realizar con ese objeto.

Un ejemplo de objeto son las listas en Python, representadas por el tipo `list`. El tipo `list` define las operaciones (funcionalidad)  que se pueden realizar con las listas, como por ejemplo `append`, `pop`, `remove`, etc. 
Estas operaciones pueden ser invocadas con el operador punto, de la forma: `lista.pop()`.

## 1. Las clases

Para definir nuevos tipos de datos, usamos `clases`. Mediante una clase podemos indicar las `propiedades` (atributos o  características) que van a tener los elementos del nuevo tipo de datos y las `operaciones` ( métodos) que vamos a poder realizar con cada uno de los elementos pertenecientes a este tipo de datos.

En términos de programación, uno empieza definiendo una __clase__ y luego procede a crear __objetos__ de dicha clase.


<img src="./images/clase.png" alt="clase" style="width: 300px;"/>


A los elementos de una clase se les llama `objetos` o `instancias` de la clase. En ejecución, se tendrán variables que apuntarán a los objetos creados. 

### Primer ejemplo. La clase `Vehículo`

Python define sus clases con la palabra reservada __class__ seguida del  nombre de la clase.

Para crear objetos en Python, se usa una función llamada `__init__`. Esta función o `método` es el inicializador del objeto y se ejecuta automáticamente (lo veremos más adelante). Se llama `constructora de objetos`.

````python
# versión 1
class Vehiculo:
     
    def __init__(self) -> None:
        self.color = 'Blanco'
        self.ancho = 2.5
        self.velocidad = 0
        self.plazas = 4
````

````python
coche_1 = Vehiculo()
coche_2 = Vehiculo()
````

````python
coche_1.plazas
````

````python
coche_1.color
````

````python
print(coche_1)
````

````python
# versión 2
class Vehiculo:
    def __init__(self, c:str ='Blanco', a:float = 2.5, v:float = 0, p:int = 4):
        
        self.color = c
        self.ancho = a
        self.velocidad = v
        self.plazas = p
    
    def cambiarColor(self, nuevo_col:str) -> None:
        """
        >>> b = Vehiculo()
        >>> b.cambiarColor('Rojo')
        Traceback (most recent call last):
          ...
        NameError: name 'nuevo_color' is not defined
        """
        if nuevo_col == 'Rojo':
            raise ValueError(f'El color  {nuevo_color} no está permitido')
        self.color = nuevo_col

    def parar(self) -> None:
        """
        Precondition
        ------------
        
        """
        self.velocidad = 0
````

````python
import doctest
doctest.testmod(verbose=False)
````

````python
coche_1 = Vehiculo('Rojo')
coche_2 = Vehiculo(c='Verde', a=3, v=100)
coche_2.cambiarColor('Azul')
````

El método `__init__` es un método especial (este método se llama `constructora` de la clase), el cual se ejecuta automáticamente al crear el objeto.  

Para crear un objeto de la clase, simplemente hay que invocar a la clase como si fuera una función, pasando los argumentos que defina el método `__init__`. El valor de retorno será el objeto recién creado.

El parámetro `self` es obligatorio en todos los __métodos de instancia__ (métodos que se ejecutan sobre un objeto), __el cual no se pasa cuando se  llama__ al método. Contiene una referencia al propio objeto, por lo que se utiliza para acceder a cualquier atributo o método del propio objeto, como en el caso del método `__init__`, donde accede a los atributos del propio objeto para asignarle los valores pasados por parámetro.

Una vez creado el objeto, es posible invocar a los métodos de instancia que define la clase sobre ese objeto. El código de los métodos está declarado en la clase y actúa sobre los atributos concretos del objeto.

### Atributos

Por convención, todos los atributos de una clase se inicializan en el método `__init__`. Es una buena práctica asignar un valor inicial a todos los atributos de un objeto. Esto evitará futuros errores cuando se quiera acceder a un atributo no inicializado.

#### Acceso a los atributos dentro de la clase

Para hacer referencia a un atributo dentro de la clase, se debe usar __self__ como calificador, (__self.color__, __self.plazas__, ...).

#### Acceso a los atributos desde fuera de la clase

El acceso a los atributos del objeto desde fuera de la clase se realiza nombrando al objeto, seguido de un punto, y del nombre del atributo  (__coche_1.color__, __coche_1.plazas__, ...).

Normalmente los atributos de los objetos son públicos. El concepto de `atributo privado`, que no puede accederse excepto desde dentro de un objeto, no existe en Python. Sin embargo, hay una convención que se sigue en la mayoría del código Python: un nombre prefijado con un guión bajo (por ejemplo, `_color`) debería tratarse como una parte no pública de la API. Debería considerarse un detalle de implementación y que está sujeto a cambios sin aviso.


#### El atributo self

Es el primer argumento de cada método de una clase (incluido `__init__`), siempre es una referencia al objeto actual de la clase. En el método `__init__`, `self` hace referencia al objeto recién creado; y en otros métodos de la clase, se refiere al objeto cuyo método ha sido llamado.

Aunque es necesario especificar `self` de forma explícita cuando se define un método, no se especifica al invocar el método; Python lo añadirá de forma automática.



### Métodos

El acceso a los  métodos se realiza nombrando al objeto, seguido de un punto, y del nombre del atributo o del método. 

La función `dir()` obtiene los métodos que posee un objeto:



````python
coche_1   = Vehiculo("Rojo", 2, 0, 2)
coche_2 = Vehiculo("Verde", 1.8, 100, 4)
# actualizamos atributos
coche_1.plazas = 5
coche_2.ancho = 3
# invocamos métodos
coche_2.parar()
coche_1.cambiarColor('Azul')
````

# 2. Métodos `str`, `repr`


Python cuenta con una serie de métodos especiales. Estos métodos pueden ser invocados manualmente o pueden ser invocados automáticamente cuando se usa una sintaxis específica, con el fin de emular cierto tipo de funcionalidades. Estos métodos se invocan de manera automática.

## \_\_str\_\_ 

El método especial `__str__` es llamado de forma automática cuando se ejecuta la función `print` de un objeto de la clase. Este método devuelve una cadena de caracteres (un string) con una representación del objeto legible y fácil de interpretar por el usuario. Es una __representación informal__ del objeto en formato string.

Veamos un ejemplo:

````python
class Vehiculo:

    def __init__(self, c ='Blanco', a = 2.5, v = 0, p = 4):
        self.color = c
        self.ancho = a
        self.velocidad = v
        self.plazas = p
        
    def __str__(self):
        return f'Color {self.color}, ancho {self.ancho}'
    
````

````python
coche_1 = Vehiculo('Azul', 3, 100, 4)
````

````python
coche_1
````

````python
print(coche_1)
````

## \_\_repr\_\_ 

El objetivo del método `__repr__` es devolver una cadena que muestre una representación interna del objeto. Es decir, el método `__repr__` debe devolver una expresión que permita volver a crear el objeto con los mismos valores. Es una __representación formal__ del objeto en formato string.

Podemos definir nuestro método especial `__repr__` en nuestra clase así:

````python
class Vehiculo:

    def __init__(self, c ='Blanco', a = 2.5, v = 0, p = 4):
        self.color = c
        self.ancho = a
        self.velocidad = v
        self.plazas = p
        
    def __repr__(self):
        return f'Vehiculo({self.color}, {self.ancho}, {self.velocidad}, {self.plazas})'
    
````

````python
coche_1 = Vehiculo('Azul', 3, 100, 4)
````

````python
print(coche_1)
````

# 3. Miembros estáticos y métodos de clase

En Python, los miembros estáticos son aquellos atributos y métodos que pertenecen a la clase __en lugar de a una instancia específica__. 

Vamos a ver cómo se definen y usan atributos estáticos y métodos estáticos (y también los métodos de clase, que están relacionados).

## Atributos estáticos (de clase)

Son variables que pertenecen a la clase en sí y no a una instancia (objeto) en particular.

Se definen directamente en la clase, fuera de cualquier método. Se comparten entre todas las instancias de la clase.

__Ejemplo 1:__

````python
class Contador:
    # Atributo estático (de clase)
    total_instancias = 0    # variable

    def __init__(self):
        # Accedemos al atributo de clase y lo incrementamos
        Contador.total_instancias += 1

# Uso
a = Contador()
b = Contador()
c = Contador()

print(Contador.total_instancias)  # 
print(a.total_instancias)         #  (acceso desde instancia, pero es el mismo valor)
````

> Si se modificar el atributo de clase desde un objeto específico (`a.total_instancias = ...`), Python no modifica la variable de clase, sino que crea dinámicamente un nuevo atributo de instancia en `a` que oculta al atributo de clase.

__Ejemplo 2:__

````python
class Matricula:
    
    PREFIJO = 'ES'    # actua como una constante de clase

    def __init__(self, valor: int):
        # Accedemos al atributo de clase y lo incrementamos
        self.cod = f'{Matricula.PREFIJO}-{valor}'
        
    def __repr__(self):
        return self.cod

````

````python
# Uso
a = Matricula(3356)
b = Matricula(4545)
c = Matricula(4560)

a, b, c
````

````python
a.PREFIJO
````

Aunque `PREFIJO` es un atributo de clase, las instancias pueden acceder a él porque Python busca primero en la instancia, y si no lo encuentra, busca en la clase (y luego en las clases padre, si hay herencia).

````python
a.PREFIJO    # No usar de esta forma
````

__OJO!__: Si hacemos `a.PREFIJO = 'FR'`, estaremos creando un atributo de instancia llamado `PREFIJO` en el objeto `a` (distinto al de la clase), que ocultará el de la clase solo para esa instancia.

````python
# Ahora modificamos desde la instancia
a.PREFIJO = 'FR'

print(a.PREFIJO)       
print(b.PREFIJO)       
print(Matricula.PREFIJO)  
````

Si queremos garantizar que no se modifique el valor de una contante, podemos usar `Final`. 

````python
from typing import Final

class Matricula2:
    PREFIJO: Final[str] = 'ES'

    def __init__(self, valor: int):
        # Accedemos al atributo de clase y lo incrementamos
        self.cod = f'{Matricula2.PREFIJO}-{valor}'
        
    def __repr__(self):
        return self.cod

a = Matricula2(4587)
a
````

````python
Matricula2.PREFIJO = 'ED'
````



## Métodos estáticos y de clase

---

En python existen varios tipos de métodos. 

### Métodos de instancia

Los `métodos de instancia` son métodos que reciben como parámetro la palabra `self`, que representa un objeto (o una instancia). Estos métodos solo pueden ser llamados a partir de un objeto de la clase.

Ejemplos de  métodos de instancia son los métodos llamados `especiales`, como son `__init__`, `__eq__`, `__str__`, etc. También son métodos de instancia todas las funciones que reciben `self` como parámetro.

````python
class Vehiculo:

    def __init__(self, c ='Blanco', a = 2.5, v = 0, p = 4):
        self._color = c
        self._ancho = a
        self._velocidad = v
        self._plazas = p

    def __str__(self):    # método de instancia
        return f'Color {self._color}, ancho {self._ancho}' 
    
    def cambiarColor(self, nuevo_col):      # método de instancia
         self._color = nuevo_col

    def parar(self):                        # método de instancia
        self._velocidad = 0
````

Para manipular o interactuar con la clase o atributos estáticos, Python ofrece dos decoradores: `@classmethod` vs `@staticmethod`.

### Métodos estáticos

Los métodos estáticos tienen la característica de que __no necesitan estar vincuados a un objeto__ de la clase para poder ser llamados. 

* Por tanto, estos métodos se corresponden con funciones que no reciben el  parámetro `self` en la definición. No dependen de los atributos de la clase.

*  <span style="color:red;">Se invocan con el nombre de la clase</span>.

* Se usan para definir **funciones auxiliares** relacionadas con la clase, pero que **no necesitan acceder a atributos de la clase ni instancias**.

Estos métodos son marcados con un decorador llamado `@estaticmethod`. Este decorador se escribe justo antes de la definición del método estático y sirve para indicar que es un método estático.

````python
class Vehiculo:

    def __init__(self, c ='Blanco', a = 2.5, v = 0, p = 4):
        self._color = c if Vehiculo.validar_color(c) else 'Blanco'
        self._ancho = a
        self._velocidad = v
        self._plazas = p
    
    def __str__(self):    # método de instancia
        return f'Color {self._color}, ancho {self._ancho}' 

    def __repr__(self):   # método de instancia
        return f'Vehiculo({self._color}, {self._ancho},  {self._velocidad}, {self._plazas})'

    def parar(self):     # método de instancia
        self._velocidad = 0
    
    @staticmethod
    def validar_color(n):
        return n in ['Rojo', 'Blanco', 'Azul']


   
````

````python
coche_1 = Vehiculo('Morado')
print(coche_1)
````

````python
coche_2 = Vehiculo('Azul')
print(coche_2)
````

Los métodos estáticos no están vinculados a un objeto de la clase, pero sí __están vincuados a la clase__, por lo que se llaman anteponiendo el nombre de la clase. 

Pueden ser llamados desde fuera de la clase o desde un método de instancia.



### Métodos de clase

Un método de clase recibe como primer argumento la clase misma (__cls__, no self). Útil para **acceder o modificar atributos de clase** o **crear constructores alternativos**.



````python
class Vehiculo:
    # atributos de clase
    num_coches = 0

    def __init__(self, c ='Blanco'):
        self._color = c 
        Vehiculo.num_coches += 1
        
    def __repr__(self):   # método de instancia
        return f'Vehiculo({self._color})'
    
    @classmethod
    def contar_coches(cls):
        # Recibe la clase (cls) como primer parámetro
        return cls.num_coches

    @classmethod
    def crear_coche(cls, color):
        return cls(color)

c1 = Vehiculo()
c2 = Vehiculo.crear_coche('verde')
c1 , c2 , Vehiculo.num_coches
````


# 4. Otros métodos especiales

---

Python cuenta con una serie de métodos especiales. Estos métodos pueden ser invocados manualmente o pueden ser invocados automáticamente cuando se usa una sintaxis específica, con el fin de emular cierto tipo de funcionalidades. Estos métodos se invocan de manera automática.

##  \_\_len\_\_


Es un método especial que se invoca cuando llamamos a “len(objeto)” y busca emular la funcionalidad de la función interna len(). 

Funciona sobre cualquier objeto del que se pueda pensar razonablemente que tiene longitud o tamaño. 

* La función __len()__ de una cadena es su número de caracteres; 
*  __len()__ para un diccionario es su número de claves;
* __len()__  de una lista o tupla es su número de elementos.

Volvamos a la clase  __Vehiculo__. podemos crear nuestro método especial __len__ así:

````python
class Vehiculo:
    def __init__(self, aColor, aAncho, aVel, aPlazas):
        # los objetos de la clase vehículo tienen 4 atributos
        self._color = aColor
        self._ancho = aAncho
        self._velocidad = aVel
        self._plazas = aPlazas
        
    def __str__(self): 
        texto = "Color: " + self._color + '\n' + "Plazas: " \
                + repr(self._plazas) + '\n' + 'Velocidad: ' + repr(self._velocidad)
        return texto
    
    def __len__(self): 
        return self._plazas         


    def __eq__(self, otro): 
        return self._color == otro._color and self._ancho == otro._ancho and self._plazas == otro._plazas
        
# creación de objetos        
coche_1  = Vehiculo("Rojo", 3, 0, 7)
coche_2  = Vehiculo("Rojo", 3, 0, 9)
````

Para probar el método, basta con invocar la función len():

````python
len(coche_1)
````

## Comparaciones "ricas"

Estos  métodos se definen con dos argumentos de entrada (dos objetos) y  se invocan cuando realizamos operaciones de comparación con los objetos de una clase. Son los métodos de comparación enriquecida:

* la operación  `x<y` llama al método especial  `x.__lt__(y)`,
* `x<=y ` llama a  `x.__le__(y) `,
* `x==y` llama a  `x.__eq__(y)`,
* `x!=y`  llama a  `x.__ne__(y)`, 
*  `x>y`   llama a  `x.__gt__(y)`, 
* `x>=y ` llama a  `x.__ge__(y)`.

Por ejemplo, en la clase `Vehiculo` podemos decir que dos coches son iguales si tienen el mismo número de plazas:



````python
class Vehiculo:
    def __init__(self, aColor, aAncho =2.3):
        # los objetos de la clase vehículo tienen 4 atributos
        self.color = aColor
        self.ancho = aAncho
        self.velocidad = 0
        self.plazas = 4
        
    def __str__(self): 
        texto = "Color: " + self.color + '\n' + "Plazas: " \
                + repr(self.plazas) + '\n' + 'Velocidad: ' + repr(self.velocidad)
        return texto
    
    def __len__(self): 
        return self._plazas
    
    def __eq__(self, otro): 
        """
        Dos coches son iguales si tienen el mismo número de plazas
        """
        return self.plazas == otro.plazas and self.color == otro.color and self.ancho == otro.ancho
    
````

````python
# creación de objetos        
coche_1  = Vehiculo("Verde", 3)
coche_2 = Vehiculo("Verde", 3)
````

````python
coche_1 == coche_2
````

````python
# creación de objetos        
coche_3  = Vehiculo("Verde", 3)
coche_4 = Vehiculo("Rojo", 3)
````

````python
coche_3 == coche_4
````

No es necesario definir todos los métodos especiales. Podemos usar el decorador de clase `@total_ordering` , el cuál proporcionará el resto.  

Con este decorador, solo será necesario definir `__eq__` y uno de los 4 siguientes: `__lt__`, `__le__`, `__gt__`,  `__ge__`.

https://docs.python.org/3/library/functools.html#functools.total_ordering



````python
import functools
@functools.total_ordering

class Vehiculo:
    def __init__(self, aColor, aAncho =2.3,  v= 0, p = 4):
        # los objetos de la clase vehículo tienen 4 atributos
        self.color = aColor
        self.ancho = aAncho
        self.velocidad = v
        self.plazas = p
        
    def __str__(self): 
        texto = "Color: " + self.color + '\n' + "Plazas: " \
                + repr(self.plazas) + '\n' + 'Velocidad: ' + repr(self.velocidad)
        return texto
    
    def __len__(self): 
        return self._plazas
    
    def __eq__(self, otro): 
        """
        Dos coches son iguales si tienen el mismo número de plazas
        """
        return self.plazas == otro.plazas and self.color == otro.color and self.ancho == otro.ancho
    
    
    def __lt__(self, otro):        
        return self.plazas < otro.plazas
    
    # resto de métodos
    def cambiarColor(self, nuevo_col):
         self._color = nuevo_col

    def parar(self):
        self._velocidad = 0
        
````

Ahora podemos probar los métodos:

````python
# creación de objetos        
coche_1  = Vehiculo("Rojito", 1, 0, 2)
coche_2 = Vehiculo("Verde", 2, 0, 7)
coche_3  = Vehiculo("Azul", 3, 0, 7)
````

````python
coche_1 < coche_2
````

````python
coche_1 >= coche_3
````

## Hash

Algunos objetos en Python tienen asignado un código interno que los identifica de manera unívoca. Este código se llama código `hash` y es un entero de 32 bits.

Para conocer el código `hash` de una variable, es necesario ejecutar la función `hash` sobre dicha variable. 

````python
# código hash de un número entero
# Los números enteros son objetos de la clase int
a = 7
hash(a)
````

````python
# código hash de un string
s1 = 'Hola mundo'
s2 = 'Hola mundo'
print(hash(s1))
print(hash(s2))
````

````python
id(s1), id(s2)
````

Pero no todos los objetos tienen código `hash`. No todos los objetos son __Hashables__. Se dice que un objeto es Hashable si su código hash es siempre el mismo en cualquier momento de la vida de dicho objeto (inmutable).

* En el caso de los **diccionarios**, es imprescindible que las claves sean objetos `hashables`.
* En el caso de **conjuntos**, los elementos también tienen que ser `hashables`.

En el caso de las **listas**, como son un tipo mutable, no tienen código `hash`, por tanto no son hashables. Las listas pueden cambiar durante la ejecución de un programa. Esta es una de las razones por las que no pueden ser claves de un diccionario.

````python
m1 = [1, 2, 3]
hash(m1)
````

En el caso de las clases definidas por el usuario, los objetos de la clase son hashables por defecto, todos tienen un código `hash` diferente. 

Veamos un ejemplo:

````python
class Nif:
    def __init__(self, numero , letra):
        self._numero = numero
        self._letra = letra
    
    def __repr__(self):
        texto = str(self._numero) + str(self._letra)
        return texto

b1 = Nif(80001000, 'N')
b2 = Nif(80001000, 'N')
print(b1, ',', b2)
hash(b1), hash(b2)
````

Los objetos de la clase `Nif` son `hashables`. Tienen los mismos valores para sus parámetros, pero no tienen el mismo código `hash`.

Si quiero definir una variable como un diccionario donde las claves son objetos de la clase Nif, haría lo siguiente:

````python
biblioteca = {}
biblioteca[b1] = 'Alberto'
biblioteca[b2] = 'Ana'
print(biblioteca)
````

Como podemos observar, las claves del diccionario son las mismas; no 'parece' que estemos obteniendo el resultado deseado. Para el usuario, los nif's `b1` y `b2` son el mismo nif, por lo que no se deberían repetir en el diccionario. Lo que ocurre, es que tienen distinto código hash, por lo que para Python representan claves diferentes.

Para arreglar este comportamiento, y sólo si queremos utilizar objetos de nuestras clases como claves de un diccionario, es necesario implementar los  métodos especiales `__hash__` y `__eq__`. La implementación debe devolver el mismo código `hash` para todos los objetos que se consideren iguales.

````python
class Nif:
    def __init__(self, numero , letra):
        self._numero = numero
        self._letra = letra
    
    def __repr__(self):
        texto = str(self._numero) + str(self._letra)
        return texto
    
    def __eq__(self, otro):
        return self._numero == otro._numero and self._letra == otro._letra
    
    def __hash__(self):
        return hash(str(self._numero) + self._letra)
````

````python
b1 = Nif(80001000, 'N')
b2 = Nif(80001000, 'N')
print(b1, ',', b2)
#print(hash(b1), hash(b2))

biblioteca = {}
biblioteca[b1] = 'Ana'
biblioteca[b2] = 'Alberto'
print(biblioteca)
````

Es importante que se cumpla que todos los objetos que se comparan con `==` devuelvan el mismo valor de hash. Si un objeto no define `__eq__` tampoco debería definir `__hash__`. Si define `__eq__` pero no `__hash__`, sus bjetos no podrán utilizarse como claves en diccionarios.

## Operadores aritméticos

Python permite que los operadores aritméticos (+, -, *, etc) puedan ser definidos para nuestras clases.

| Operación | Método Directo | Método Inverso (*Reverse*) | Método In-place (`+=`, `-=`) |
| --- | --- | --- | --- |
| **`x + y`** | `x.__add__(y)` | `y.__radd__(x)` | `x.__iadd__(y)` |
| **`x - y`** | `x.__sub__(y)` | `y.__rsub__(x)` | `x.__isub__(y)` |
| **`x * y`** | `x.__mul__(y)` | `y.__rmul__(x)` | `x.__imul__(y)` |
| **`x / y`** | `x.__truediv__(y)` | `y.__rtruediv__(x)` | `x.__itruediv__(y)` |
| **`x // y`** | `x.__floordiv__(y)` | `y.__rfloordiv__(x)` | `x.__ifloordiv__(y)` |
| **`x % y`** | `x.__mod__(y)` | `y.__rmod__(x)` | `x.__imod__(y)` |
| **`x ** y`** | `x.__pow__(y)` | `y.__rpow__(x)` | `x.__ipow__(y)` |
| **`-x` (Negativo)** | `x.__neg__()` | N/A | N/A |
| **`+x` (Positivo)** | `x.__pos__()` | N/A | N/A |

Python primero intenta llamar al método directo del objeto de la izquierda (ej. `x.__add__(y)`). Si dicho método no está definido o devuelve el valor especial `NotImplemented`, entonces Python intenta ejecutar el método inverso del objeto de la derecha (`y.__radd__(x)`). Si este también devuelve `NotImplemented` o no existe, se lanza un `TypeError`.

Asimismo, para el operador unario `-x`, el método especial correspondiente es `x.__neg__()` (no recibe el parámetro y, ya que solo actúa sobre un operador).

La lista completa está en el enlace: https://docs.python.org/3/reference/datamodel.html#emulating-numeric-types

````python
from typing import Self

class Vector2D:
    def __init__(self, x: float, y: float) -> None:
        self.x: float = x
        self.y: float = y

    #  Suma Directa: v1 + v2 
    def __add__(self, other: Self) -> Self:
        return Vector2D(self.x + other.x, self.y + other.y)

    #  Operador Unario Negativo: -v1
    def __neg__(self) -> Self:
        return Vector2D(-self.x, -self.y)

    # 3 Producot escalar : v1 * v2
    def __mul__(self, other: Self) -> float:
        return self.x * other.x + self.y * other.y

    # 4. Multiplicación Inversa: numero * v1
    def __rmul__(self, scalar:float) -> Self:
        return Vector2D(self.x * scalar, self.y * scalar)

    def __repr__(self) -> str:
        return f"Vector2D({self.x}, {self.y})"


# Uso de la clase y prueba de operadores
v1 = Vector2D(2, 3)
v2 = Vector2D(4, 1)

print(v1 + v2)      #  Ejecuta v1.__add__(v2)
print(v1 * v2)      #  Ejecuta v1.__mul__(v2)
print(10 * v1)      #  Ejecuta v1.__rmul__(3)
print(-v1)          #  Ejecuta v1.__neg__()
````

## Instrucción `with`

El uso de la instrucción `with` es la forma más limpia de garantizar la gestión de recursos (archivos, conexiones a bases de datos, ...).

Por ejemplo para ficheros:

```python
      with open ( filename , 'w') as fl :
        do_something(fl)
```

Para que esto funcione, la clase debe implementar dos métodos especiales: `__enter__(self)`y `__exit__(self, exc_type, exc_value, traceback)`.

## `__enter__(self)`

Se ejecuta automáticamente cuando se entra en el bloque `with`. Suele devolver `self`, aunque puede devolver cualquier otro objeto.
El valor que devuelve se asigna a la variable después de `as`.

```python
        def __enter__(self):
            print("Entrando")
            return self


        with MiClase() as obj:
            print(obj)
```

## `__exit__(self, exc_type, exc_value, traceback)`

Se ejecuta automáticamente al salir del bloque `with`, tanto si todo ha ido bien como si se ha producido una excepción.

```python
        def __exit__(self, exc_type, exc_value, traceback):
            print("Saliendo")
```

Recibe tres parámetros relacionados con posibles excepciones:

* `exc_type`: tipo de excepción (ValueError, ZeroDivisionError, etc.).
* `exc_value`: objeto de la excepción (el mensaje de error).
* `traceback`: información sobre dónde ocurrió el error.

````python
from types import TracebackType
from typing import Optional
from sqlalchemy import create_engine
from sqlalchemy.engine import Connection, Engine

class DB:
    """Administrador de contexto para gestionar el ciclo de vida 
    de una conexión a base de datos mediante SQLAlchemy.

    :param method: Cadena de conexión (URL) a la base de datos.
    :type method: str
    """
    def __init__(self, method: str) -> None:
        self.engine: Engine = create_engine(method)
        self.con: Optional[Connection] = None

    def __enter__(self) -> Connection:
        """Abre la conexión y la retorna para ser usada en el bloque `with`.

        :return: Objeto de conexión activo de SQLAlchemy.
        :rtype: Connection
        """
        self.con = self.engine.connect()
        return self.con

    def __exit__(
        self,
        exc_type: Optional[type[BaseException]],
        exc_val: Optional[BaseException],
        exc_tb: Optional[TracebackType]
    ) -> bool:
        """Cierra la conexión activa al salir del bloque `with`."""
        if self.con is not None:
            self.con.close()
        


def fun() -> None:
    dburl = 'mysql+pymysql://user:pwd@localhost:6606/db'
    
    # Se pasa la variable corregida (dburl)
    with DB(dburl) as con:
        do_something(con)
````

Si en cualquier momento se produce un error en `do_something`, no se van a dejar conexiones abiertas.

## Dataclass

El decorador` @dataclass` sirve para facilitar la creación de clases que principalmente almacenan datos.

Sin `@dataclass`, cuando creamos una clase con atributos, debemos escribir manualmente métodos como:

* __init__() para inicializar los atributos.
* __repr__() para mostrar el objeto de forma legible.
* __eq__() para comparar objetos.

````python
from dataclasses import dataclass
from typing import ClassVar

@dataclass
class InventoryItem:
    """Clase para almacenar información de un producto."""

    # Atributos de clase: se deben anotar obligatoriamente con ClassVar
    total_productos: ClassVar[int] = 0

    # Atributos de instancia
    name: str
    unit_price: float
    quantity_on_hand: int = 0

    def total_cost(self) -> float:
        return self.unit_price * self.quantity_on_hand
````

El decorador crea automáticamente un método parecido a:

```python
        def __init__(self, name: str,
                        unit_price: float,
                        quantity_on_hand: int = 0):

            self.name = name
            self.unit_price = unit_price
            self.quantity_on_hand = quantity_on_hand
```

para crear objetos:

````python
item = InventoryItem("Teclado", 25.5, 10)
item
````

También genera automáticamente `__repr__`.

La principal ventaja es reducir código repetitivo y hacer las clases más claras y fáciles de mantener.

Si es necesario validar o calcular un atributo justo después de que el ` __init__` automático haya asignado los valores, se utiliza el método especial `__post_init__`:

````python
from dataclasses import dataclass, field
from typing import ClassVar

@dataclass
class InventoryItem:
    """Clase para almacenar información de un producto."""
    # Atributos de instancia
    name: str
    unit_price: float
    impuesto: int = field(init=False)  # No se pide en el __init__

    def __post_init__(self) -> None:
            """Se ejecuta automáticamente AL FINAL del __init__ generado."""
            if self.unit_price < 0:
                raise ValueError("El precio no puede ser negativo.")
            self.impuesto = self.unit_price * 0.21

    def total_cost(self) -> float:
        return self.unit_price + self.impuesto
````
