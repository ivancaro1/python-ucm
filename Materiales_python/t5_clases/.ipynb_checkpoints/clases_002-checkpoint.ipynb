
# Métodos especiales


Python cuenta con una serie de métodos especiales. Estos métodos pueden ser invocados manualmente o pueden ser invocados automáticamente cuando se usa una sintaxis específica, con el fin de emular cierto tipo de funcionalidades. Estos métodos se invocan de manera automática.

## \_\_str\_\_ 

El método especial `__str__` es llamado de forma automática cuando se ejecuta la función `print` de un objeto de la clase. Este método devuelve una cadena de caracteres (un string) con una representación del objeto legible y fácil de interpretar por el usuario. Es una __representación informal__ del objeto en formato string.

Veamos un ejemplo:

```python
class Vehiculo:

    def __init__(self, c:str ='Blanco', a:float = 2.5, v:float = 0, p:int = 4) -> None:
        self.color = c
        self.ancho = a
        self.velocidad = v
        self.plazas = p
        
coche_1 = Vehiculo('Azul', 3, 100, 4)


print(coche_1)    
coche_1
```

En ambos casos, la información que se ha mostrado es el nombre de la clase al que pertenece el objeto `coche_1` y la posición de memoria donde se encuentra almacenado. Pero no vemos las características del objeto, nombre, etc.

```python
class Vehiculo:

    def __init__(self, c:str ='Blanco', a:float = 2.5, v:float = 0, p:int = 4) -> None:
        self.color = c
        self.ancho = a
        self.velocidad = v
        self.plazas = p
        
    def __str__(self) -> str:
        return f'Color {self.color}, ancho {self.ancho}'
    
```

```python
coche_1 = Vehiculo('Azul', 3, 100, 4)
```

```python
coche_1
```

```python
print(coche_1)
```

## \_\_repr\_\_ 

El objetivo del método `__repr__` es devolver una cadena que muestre una representación interna del objeto. Es decir, el método `__repr__` debe devolver una expresión que permita volver a crear el objeto con los mismos valores. Es una __representación formal__ del objeto en formato string.

Podemos definir nuestro método especial `__repr__` en nuestra clase así:

```python
class Vehiculo:

    def __init__(self, c:str ='Blanco', a:float = 2.5, v:float = 0, p:int = 4) -> None:
        self.color = c
        self.ancho = a
        self.velocidad = v
        self.plazas = p
        
    def __repr__(self) -> str:
        return f'Vehiculo({self.color}, {self.ancho}, {self.velocidad}, {self.plazas})'
    
```

```python
coche_1 = Vehiculo('Azul', 3, 100, 4)
```

```python
print(coche_1)
```

```python
coche_1
```

## Implementando ambos métodos

Cuando se ejecuta `print(c)`, el intérprete de Python llama al método especial `__str__`. Si este método no existe, llama al método especial `__repr__`. Y si no existe ninguno de los dos, nos devuelve la cadena `< descripción_del_objeto >`.

<img src="./images/str.png" alt="while" style="width: 600px;"/>

```python
class Vehiculo:

    def __init__(self, c:str ='Blanco', a:float = 2.5, v:float = 0, p:int = 4) -> None:
        self.color = c
        self.ancho = a
        self.velocidad = v
        self.plazas = p
        
    def __str__(self) -> str:
        return f'Color {self.color}, ancho {self.ancho}'
    
                    
    def __repr__(self) -> str:
        return f'Vehiculo({self.color}, {self.ancho}, {self.velocidad}, {self.plazas})'
    

```

```python
coche_1 = Vehiculo('Azul', 3, 100, 4)
```

Pero también puedo consultar el valor del objeto sin necesidad de hacer `print`:

```python
print(coche_1)
```

```python
coche_1
```
