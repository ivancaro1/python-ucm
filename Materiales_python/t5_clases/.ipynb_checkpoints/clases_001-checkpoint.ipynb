#  Clases y Objetos de usuario


## Un poco de terminología

En el contexto de la programación, las `clases` son un mecanismo diseñado para construir tipos de datos similares a los objetos de la vida real. Si nuestros programas usan clases, hablamos de __programación orientada a objetos__.

Los lenguajes de programación orientados a objetos (Python lo es), permiten definir nuevos tipos de datos cuando los tipos proporcionados por el lenguaje no cubren todas las necesidades de la aplicación que estamos diseñando.

Recordar que los tipos de datos que hemos visto hasta ahora son:

* Objetos de tipo escalar: `int`, `float`, `bool`, `None`.
* Objetos de tipo no escalar: `list`,  `tuple`, `str`, `dict`.


Los __objetos__ son un mecanismo de los lenguajes de programación para unir datos y  funcionalidad: 

__Datos__:

* Un determinado objeto tiene un conjunto de características (denominadas __atributos__), cada una de las cuales tiene un valor. Estas características indican cuál es su estado actual.  
* Los objetos pueden tener una cantidad arbitraria de atributos de cualquier tipo.

__Funcionalidad:__

* Los objetos responden a un conjunto de operaciones (funciones ó __métodos__).

En Python, el núcleo de cosas que pueden manipular los programas son los objetos. Cada objeto tiene un tipo donde se definen las operaciones que se pueden realizar con ese objeto.

Un ejemplo de objeto son las listas en Python, representadas por el tipo `list`. El tipo `list` define las operaciones (funcionalidad)  que se pueden realizar con las listas, como por ejemplo `append`, `pop`, `remove`, etc. 
Estas operaciones pueden ser invocadas con el operador punto, de la forma: `lista.pop()`.

## Las clases

Para definir nuevos tipos de datos, usamos `clases`. Mediante una clase podemos indicar las `propiedades` (atributos o  características) que van a tener los elementos del nuevo tipo de datos y las `operaciones` ( métodos) que vamos a poder realizar con cada uno de los elementos pertenecientes a este tipo de datos.

En términos de programación, uno empieza definiendo una __clase__ y luego procede a crear __objetos__ de dicha clase.


<img src="./images/clase.png" alt="clase" style="width: 300px;"/>


A los elementos de una clase se les llama `objetos` o `instancias` de la clase. En ejecución, se tendrán variables que apuntarán a los objetos creados. 

### Primer ejemplo. La clase `Vehículo`

Python define sus clases con la palabra reservada __class__ seguida del  nombre de la clase.

Para crear objetos en Python, se usa una función llamada `__init__`. Esta función o `método` es el inicializador del objeto y se ejecuta automáticamente (lo veremos más adelante). Se llama `constructora de objetos`.

```python
# versión 1
class Vehiculo:     
    def __init__(self):
        self.color = 'Blanco'
        self.ancho = 2.5
        self.velocidad = 0
        self.plazas = 4
```

```python
coche_1 = Vehiculo()
coche_2 = Vehiculo()
```

```python
coche_1
```

```python
coche_1.plazas = 3

coche_1.plazas
```

```python
coche_1.color
```

```python
# versión 2
class Vehiculo:
    def __init__(self, c:str= 'Blanco'                 , 
                 a:float = 2.5, v:float = 0, p:int = 4) -> None: 
        self.color = c
        self.ancho = a
        self.velocidad = v
        self.plazas = p
    
    def cambiarColor(self, nuevo_col: str) -> None:
         self.color = nuevo_col

    def parar(self)-> None:
        self.velocidad = 0
```

```python
coche_0 = Vehiculo()
coche_1 = Vehiculo('Rojo')
coche_2 = Vehiculo(c='Verde', a=3, v=100)
```

```python
coche_1.cambiarColor('Azul')
```

El método `__init__` es un método especial (este método se llama `constructora` de la clase), el cual se ejecuta automáticamente al crear el objeto.  

Para crear un objeto de la clase, simplemente hay que invocar a la clase como si fuera una función, pasando los argumentos que defina el método `__init__`. El valor de retorno será el objeto recién creado.

El parámetro `self` es obligatorio en todos los __métodos de instancia__ (métodos que se ejecutan sobre un objeto), __el cual no se pasa cuando se  llama__ al método. Contiene una referencia al propio objeto, por lo que se utiliza para acceder a cualquier atributo o método del propio objeto, como en el caso del método `__init__`, donde accede a los atributos del propio objeto para asignarle los valores pasados por parámetro.

Una vez creado el objeto, es posible invocar a los métodos de instancia que define la clase sobre ese objeto. El código de los métodos está declarado en la clase y actúa sobre los atributos concretos del objeto.

### Atributos

Por convención, todos los atributos de una clase se inicializan en el método `__init__`. Es una buena práctica asignar un valor inicial a todos los atributos de un objeto. Esto evitará futuros errores cuando se quiera acceder a un atributo no inicializado.

#### Acceso a los atributos dentro de la clase

Para hacer referencia a un atributo dentro de la clase, se debe usar __self__ como calificador, (__self.color__, __self.plazas__, ...).

#### Acceso a los atributos desde fuera de la clase

El acceso a los atributos del objeto desde fuera de la clase se realiza nombrando al objeto, seguido de un punto, y del nombre del atributo  (__coche_1.color__, __coche_1.plazas__, ...).

Normalmente los atributos de los objetos son públicos. El concepto de `atributo privado`, que no puede accederse excepto desde dentro de un objeto, no existe en Python. Sin embargo, hay una convención que se sigue en la mayoría del código Python: un nombre prefijado con un guión bajo (por ejemplo, `_color`) debería tratarse como una parte no pública de la API. Debería considerarse un detalle de implementación y que está sujeto a cambios sin aviso.


#### El atributo self

Es el primer argumento de cada método de una clase (incluido `__init__`), siempre es una referencia al objeto actual de la clase. En el método `__init__`, `self` hace referencia al objeto recién creado; y en otros métodos de la clase, se refiere al objeto cuyo método ha sido llamado.

Aunque es necesario especificar `self` de forma explícita cuando se define un método, no se especifica al invocar el método; Python lo añadirá de forma automática.



### Métodos

El acceso a los  métodos se realiza nombrando al objeto, seguido de un punto, y del nombre del atributo o del método. 

La función `dir()` obtiene los métodos que posee un objeto:



```python
coche_1   = Vehiculo("Rojo", 2, 0, 2)
coche_2 = Vehiculo("Verde", 1.8, 100, 4)
# actualizamos atributos
coche_1.plazas = 5
coche_2.ancho = 3
# invocamos métodos
coche_2.parar()
coche_1.cambiarColor('Azul')
```

```python
print(f'El color del coche_1 es {coche_1.color}')
print(f'La velocidad del coche_2 es {coche_2.velocidad}')
```
