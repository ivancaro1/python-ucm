
# Otros métodos especiales


Python cuenta con una serie de métodos especiales. Estos métodos pueden ser invocados manualmente o pueden ser invocados automáticamente cuando se usa una sintaxis específica, con el fin de emular cierto tipo de funcionalidades. Estos métodos se invocan de manera automática.

##  \_\_len\_\_


Es un método especial que se invoca cuando llamamos a “len(objeto)” y busca emular la funcionalidad de la función interna len(). 

Funciona sobre cualquier objeto del que se pueda pensar razonablemente que tiene longitud o tamaño. 

* La función __len()__ de una cadena es su número de caracteres; 
*  __len()__ para un diccionario es su número de claves;
* __len()__  de una lista o tupla es su número de elementos.

Volvamos a la clase  __Vehiculo__. podemos crear nuestro método especial __len__ así:

```python
class Vehiculo:
    def __init__(self, aColor: str, aAncho: float, aVel: float, aPlazas: int) -> None:
        # los objetos de la clase vehículo tienen 4 atributos
        self._color = aColor
        self._ancho = aAncho
        self._velocidad = aVel
        self._plazas = aPlazas
        
    def __str__(self) -> str: 
        texto = "Color: " + self._color + '\n' + "Plazas: " \
                + repr(self._plazas) + '\n' + 'Velocidad: ' + repr(self._velocidad)
        return texto
    
    def __len__(self) -> int: 
        return self._plazas         


        
# creación de objetos        
coche_1  = Vehiculo("Rojo", 2, 0, 6)
coche_2  = Vehiculo("Rojo", 2, 0, 6)
```

Para probar el método, basta con invocar la función len():

```python
len(coche_1)
```

```python
coche_1 == coche_2
```

## Otros métodos especiales

Estos  métodos se definen con dos argumentos de entrada (dos objetos) y  se invocan cuando realizamos operaciones de comparación con los objetos de una clase. Son los métodos de comparación enriquecida:

* la operación  `x < y` llama al método especial  `x.__lt__(y)`,
* `x <= y ` llama a  `x.__le__(y) `,
* `x == y` llama a  `x.__eq__(y)`,
* `x != y`  llama a  `x.__ne__(y)`, 
*  `x > y`   llama a  `x.__gt__(y)`, 
* `x >= y ` llama a  `x.__ge__(y)`.

Por ejemplo, en la clase `Vehiculo` podemos decir que dos coches son iguales si tienen el mismo ancho:



```python
from typing import Self

class Vehiculo:
    def __init__(self, aColor, aAncho =2.3):
        # los objetos de la clase vehículo tienen 4 atributos
        self.color = aColor
        self.ancho = aAncho
        self.velocidad = 0
        self.plazas = 4
        
    def __str__(self) -> str: 
        texto = "Color: " + self.color + '\n' + "Plazas: " \
                + repr(self.plazas) + '\n' + 'Velocidad: ' + repr(self.velocidad)
        return texto
    
    def __len__(self) -> int: 
        return self._plazas
    
    def __eq__(self, otro: Self) -> bool: 
        """
        Dos coches son iguales si tienen el mismo ancho y el mismo color
        """
        return self.ancho == otro.ancho  and self.color == otro.color
    
```

```python
# creación de objetos        
coche_1  = Vehiculo("Verde", 3)
coche_2 = Vehiculo("Azul", 3)
```

```python
coche_1 == coche_2
```

```python
# creación de objetos        
coche_3  = Vehiculo("Verde", 3)
coche_4 = Vehiculo("Rojo", 2)
```

```python
coche_3 == coche_4
```

No es necesario definir todos los métodos especiales. Podemos usar el decorador de clase `@total_ordering` . Este decorador  se utiliza para reducir la cantidad de métodos que necesitas implementar en una clase personalizada para definir comparaciones de objetos. En lugar de implementar todos los métodos de comparación (`__lt__`, `__le__`, `__gt__`,  `__ge__`), solo necesitamos implementar `__eq__` y uno de los 4 siguientes: `__lt__`, `__le__`, `__gt__`,  `__ge__`, y `@total_ordering` se encargará de generar los otros.


https://docs.python.org/3/library/functools.html#functools.total_ordering



```python
from functools import total_ordering

@total_ordering
class Vehiculo:
    def __init__(self, aColor, aAncho =2.3,  v= 0, p = 4):
        # los objetos de la clase vehículo tienen 4 atributos
        self.color = aColor
        self.ancho = aAncho
        self.velocidad = v
        self.plazas = p
        
    def __str__(self) -> str: 
        texto = "Color: " + self.color + '\n' + "Plazas: " \
                + repr(self.plazas) + '\n' + 'Velocidad: ' + repr(self.velocidad)
        return texto
    
    def __len__(self)-> int: 
        return self._plazas
    
    def __eq__(self, otro: Self) -> bool: 
        """
        Dos coches son iguales si tienen el mismo número de plazas, mismo color y mismo ancho
        """
        return self.plazas == otro.plazas and 
              self.color == otro.color and self.ancho == otro.ancho
    
    
    def __lt__(self, otro: Self) -> bool:    
        return self.plazas < otro.plazas
    
    # resto de métodos
    def cambiarColor(self, nuevo_col: str) -> None:
         self._color = nuevo_col

    def parar(self) -> None:
        self._velocidad = 0
        
```

Ahora podemos probar los métodos:

```python
# creación de objetos        
coche_1  = Vehiculo("Rojito", aAncho = 1,  v = 0, p = 2)
coche_2 = Vehiculo("Verde", aAncho = 2, v = 0, p = 7)
coche_3  = Vehiculo("Azul", aAncho = 3, v = 0, p = 7)
```

```python
coche_1 < coche_2
```

```python
coche_1 >= coche_3
```
