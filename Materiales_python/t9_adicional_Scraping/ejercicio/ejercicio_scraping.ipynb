```python



def extraer_datos(url):
    # petición al servidor
    respuesta = requests.get(url)
    
    if respuesta.status_code != 200:
        return
    else:
        soup = BeautifulSoup(respuesta.text, "html.parser")
        lista_libros = []
        elementos = soup.select('.product_pod')
        for libro in elementos:
            titulo = libro.select_one('h3 a').get_text()
            precio = libro.select_one('p.price_color').get_text()
            
            d = {'titulo' : titulo, 'precio': precio.replace('Â£', '')}
            lista_libros.append(d)
        return lista_libros           
        
    
```

```python

lista_all = []
for i in range(1, 51):
    url = f'https://books.toscrape.com/catalogue/page-{1}.html'
    lista = extraer_datos(url)
    lista_all.extend(lista)
```
