# DocStrings

---

## Concepto

Todos los objetos cuentan con un atributo especial `__doc__` mediante el cual se indica el propósito y uso del objeto. Estos son los llamados __docstrings__ o cadenas de documentación.  Python ofrece un mecanismo por el cual se asocia al atributo `__doc__`  de un objeto, la primera instrucción en la definición del objeto, siempre y cuando esta instrucción sea una cadena.

En el siguiente ejemplo, la primera instrucción después de la cabecera de la función es una cadena, la cual se asignará al atributo `__doc__` de la función `gcd`.

````python
def gcd(a:int, b:int)-> int:
    """
    Calcula el máximo común divisor (MCD) de dos números enteros utilizando el algoritmo de Euclides.

    :param a: El primer número entero.
    :param b: El segundo número entero.
    
    :returns a: El máximo común divisor de `a` y `b`.
    
    :raises ValueError: Si cualquiera de los dos números es negativo.

    :example:

    >>> gcd(48, 18)
    6
    >>> gcd(54, 24)
    6
    >>> gcd(7,0)
    Traceback (most recent call last):
      ...
    ValueError: (False, 'b must be > 0')
    """
    if a <0:
        raise ValueError('a must be >= 0')
    if b <=0:
        raise ValueError(b > 0, 'b must be > 0')
    while b !=0:
        r = a % b
        a = b
        b = r
    return a 
````

Podemos imprimir el valor de dicho atributo:

````python
print(gcd.__doc__)
````

Como vemos, lo interesante de estas cadenas es que, a diferencia de los comentarios normales de Python y de los comentarios de otros lenguajes, las cadenas de documentación no se eliminan del bytecode, por lo que se pueden consultar en tiempo de ejecución, usando, por ejemplo,
la función `help` del lenguaje, o utilizando la sentencia `print` como en el ejemplo anterior.

````python
help(gcd)
````

````python
gcd?
````

##  Formatos para escribir docstring
---

###  El formato reStructuredText (reST)

El formato  reST `reStructuredText`  (creado por  David Goodger),  es un formato muy utilizado en la comunidad Python. Por supuesto, hay otros formatos, como `PEP 257 (docBlockr)` o  `Google Style Docstrings`. Este último aunque Menos formal que reST, se está haciendo muy popular por su simplicidad.

Para describir propiedades de los elementos que estamos documentando se utilizan los campos o fields. Los campos comienzan con `:`, le sigue el nombre del campo y opcionalmente sus argumentos, y se cierra de nuevo con `:`, para terminar con el cuerpo del campo.

````python
def sumar(a: int, b: int) -> int:
    """Suma dos números enteros.

    :param a: Primer operando.
    :type a: int
    :param b: Segundo operando.
    :type b: int
    :return: La suma de ambos operandos.
    :rtype: int
    """
    return a + b
````

Para indicar que utilizamos `reStructuredText` es necesario hacerlo  mediante la variable `__docformat__` en el código. Por ejemplo:

````python
'''
Supongamos que este es un script de python
Modulo para ejemplificar el uso de docstring
'''
__docformat__ = 'restructuredtext'

class Persona:
    '''Modela una persona.'''
    def __init__(self, nombre: str, edad:int)-> None:        
        '''
        Inicializador de la clase Persona.
        :param nombre: Nombre de la persona.
        :param edad: Edad de la persona
        '''
        self.nombre = nombre
        self.edad = edad
        print(self)
        
    def __str__(self)-> str:
        '''
        Imprime el nombre de la persona
        '''
        return f'Esta es la persona {self.nombre}'
               
if __name__ == '__main__' :
    juan = Persona('Juan', 26)
````

__Requisitos: precondición__

Cuando sea necesario, también se han de poner los requisitos que deben cumplir los parámetros de una función.

````python
import math

def lado_cuadrado(area: float) -> float:
    """Calcula el lado de un cuadrado conocida su área.

    :param area: El área del cuadrado.
    :type area: float

    :precondition: `area` >= 0.

    :return: El lado del cuadrado correspondiente.
    :rtype: float

    .. rubric:: Ejemplo

    >>> lado_cuadrado(2)
    1.4142135623730951
    """

    return math.sqrt(area)
````

````python
# El siguiente ejemplo fallará. Vemos que la precondición era necesaria:
print(lado_cuadrado(-2))
````

Una forma de imponer una precondición es mediante una aserción al inicio de una función:

````python
def lado_cuadrado(area: float) -> float:
    """
    .... lo de antes ----

    .. rubric:: Ejemplo

    >>> lado_cuadrado(2)
    1.4142135623730951

    >>> lado_cuadrado(-2)
    Traceback (most recent call last):
    ....
    AssertionError: El área debe ser positiva
    """
    assert area >= 0, "El área debe ser positiva"
    lado = math.sqrt(area)
    return lado
````

````python
print(lado_cuadrado(-2))
````

### El formato Google Style Docstrings


Google Style Docstrings es otro formato de documentación para Python. Como decíamos anteriormente, aunque menos formal que `reST`, se está haciendo muy popular. Es muy simple, el código es legible y fácil de mantener. 



````python
def gcd(a:int, b:int)-> int:
    """
    Divide two numbers.

    This function performs a safe division between two numeric values
    and returns the result.

    Args:
        a (int): First number. Must be greater than or equal to 0.
        b (int): Second number. Must be greater than 0.

    Returns:
        int: The greatest common divisor of `a` and `b`.

    Raises:
        ValueError: If `a` is negative.
        ValueError: If `b` is not positive.

    Examples:
        >>> gcd(48, 18)
        6
        >>> gcd(54, 24)
        6
        >>> gcd(7,0)
        Traceback (most recent call last):
          ...
        ValueError: (False, 'b must be > 0')

    Notes:
        - The function assumes integer inputs.
        - For `a = 0`, the GCD is defined as `b`.
    """
    if a <0:
        raise ValueError('a must be >= 0')
    if b <=0:
        raise ValueError(b > 0, 'b must be > 0')
    while b !=0:
        r = a % b
        a = b
        b = r
    return a 
````

````python
help(gcd)
````
