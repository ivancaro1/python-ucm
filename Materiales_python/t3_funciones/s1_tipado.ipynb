# Las anotaciones de tipos

## Anotaciones de tipos

---

Las anotaciones de tipo (_type hints_) son una forma de indicar los tipos esperados de variables, parámetros de funciones y los tipos de retorno de las funciones en Python. En otras palabras, son una manera en la que un programador puede especificar el tipo que se espera en una parte específica de un programa en Python. La mayoría de los programas en Python se escriben sin anotaciones de tipo, pero __su uso tiene muchas ventajas__. 

Vamos a empezar con una función sencilla que calcula el máximo común divisor (Algoritmo de Euclides).

````python
def gcd (a, b):
    while b !=0:
        r = a % b
        a = b
        b = r
    return a
````

La función anterior se puede anotar de la siguiente manera:

````python
def gcd (a:int, b:int)-> bool:
    while b !=0:
        r = a % b
        a = b
        b = r
    return a 
````

````python
gcd(8.2, 4)
````

* Debido a que Python no requiere que el programador especifique el tipo de una variable, la única forma de determinar el tipo de una variable sin anotaciones de tipo es mediante inspección o a través de la documentación. Esto es problemático porque hace que el código en Python sea más difícil de leer (aunque parezca lo contrario).
* Otro problema es que, debido a que Python es muy flexible, permite que el programador use la misma variable para referirse a múltiples tipos de objetos, lo cual puede derivar en errores.

Las anotaciones de tipo pueden ayudar a prevenir este estilo de programación y a reducir estos errores.

Desde que Python cuenta con anotaciones de tipo, lo consideramos un __lenguaje de tipado gradual__, lo que significa que se puede usar anotaciones de tipo cuando se desee, pero no son obligatorias. Lo cierto es que  tener anotaciones de tipo disponibles es algo positivo, algo que deberíamos aprovechar en el código.

### ¿Para qué sirven las anotaciones de tipos?

Las anotaciones de tipo, a partir de Python 3.7, __no afectan la ejecución de un programa__ en Python. En realidad, son solo sugerencias y no obligaciones.
En tiempo de ejecución, un programa en Python puede ignorar completamente sus anotaciones de tipo y romper cualquiera de las supuestas restricciones. 

* Son usadas por los analizadores estáticos de código (por ejemplo `mypy`). Las herramientas de verificación de tipos pueden evaluar las anotaciones de tipo en un programa durante el desarrollo y alertar al programador si hay llamadas inválidas a una función. Por ejemplo, una llamada a `gcd("hello", 30)` puede ser detectada antes de entrar en producción (ya que "hello" no es un valor de tipo `int`).
  
* Los utilizan los IDE (p.e PyCharm). PyCharm utiliza las anotaciones de tipos en las funciones para ofrecer una serie de beneficios durante el desarrollo de código en Python, ayudando a los desarrolladores a escribir, depurar y entender el código más fácilmente. PyCharm usa estas anotaciones para realizar verificación de tipos, autocompletado y análisis estático.
  
* Ayudan a reducir errores en los programas.

En el siguiente ejemplo, anotamos la variable donde guardaremos el resultado de la llamada a la función `gcd(4,2)`.

````python
resultado: list[str] = gcd(4, 2)
resultado
````

La anterior anotación de tipo no tiene sentido. Indica que la variable `resultado` se espera que sea una lista de cadenas (strings). Pero sabemos, por la anotación de tipo anterior, que `gcd()` devuelve un valor de tipo `int`.

Una vez más,  esta anotación de tipo errónea no tendrá ningún efecto en el funcionamiento del programa. Sin embargo, un verificador de tipos podría detectar este error del programador. El principal verificador de tipos para Python, es `mypy`.  


En proyectos gestionados con `uv`, se instala como dependencia de desarrollo mediante:
```python

```
En uv, la opción `--dev` indica que el paquete se instala como una dependencia de desarrollo (development dependency), no como una dependencia necesaria para ejecutar la aplicación (ver el fichero .tolm). 

Se ejecuta con:
```python

``` 
De esta forma, podemos detectar errores de tipos antes de ejecutar el programa.

Probar con:


`PyCharm` incluye un verificador de tipos integrado. Si editamos un programa en PyCharm con anotaciones de tipo, este señalará automáticamente los errores de tipo, ayudando a detectar los errores antes incluso de terminar de escribir una función.

¿Por qué alguien dedicaría tanto tiempo a agregar anotaciones de tipo al código si al intérprete de Python no le importan?

Las anotaciones de tipo son una buena idea por dos razones principales: 
* documentan automáticamente el código,
* permiten que un verificador de tipos compruebe la corrección de un programa antes de su ejecución

### Anotaciones para tipos no escalares

#### Listas

````python
def decompose( num: int ) -> list[int]:
    """
    Descompone un número en factores primos.
    """
    if num <= 1:
        raise ValueError ('num must be > 1')
    plst : list [int] = []
    p = 2
    while num > 1:
        if num % p == 0:
            num = num // p
            plst.append (p)
        else :
            p = p+1
    return plst
````

````python
decompose(12)
````

#### Tuplas

La siguiente función usa una tupla en su retorno y tiene un propósito claro: calcular la división, el cociente y el resto de una división.

````python
from typing import Any
def dividir_y_obtener_resto(dividendo: int, divisor: int) -> tuple[float, int, int]:

    """
    Calcula el cociente entero y el resto de la división entre dos números enteros.

    Args:
        dividendo (int): El número que se desea dividir.
        divisor (int): El número por el cual se divide el dividendo.

    Returns:
        tuple[float, int, int]: Una tupla donde el primer valor es la división, el segundo el cociente y el tercero es el resto.
    """
    if divisor == 0:
        raise ValueError("El divisor no puede ser cero.")

    division = dividendo / divisor
    cociente = dividendo // divisor  # División entera
    resto = dividendo % divisor  # Resto de la división
    return float(division), cociente, resto
````

Para tuplas y listas variables usamos:

```python
    from typing import Any

    tuple[int, ...]     # Tupla de longitud arbitraria cuyos elementos son todos enteros
    tuple[Any, ...]     # Tupla de longitud arbitraria cuyos elementos pueden ser de cualquier tipo

    list[int]      # Lista de longitud arbitraria cuyos elementos son enteros
    list[Any]      # Lista de longitud arbitraria cuyos elementos pueden ser de cualquier tipo    
```

#### Diccionarios

La función `find_word` está diseñada para buscar una lista de palabras (word) dentro de un texto (text) y devuelve un diccionario que indica las posiciones donde se encuentra cada palabra en el texto

````python
def find_word( text:str , words:list[str]) -> dict[str , list[int]]:
    word_dict: dict[str , list[int]] = {}    
    for w in words :
        word_dict[w] = []
    i = 0
    while i < len( text ):
        for w in words :
            if text [i:i+ len(w)] == w:
                word_dict[w].append(i)
        i += 1
    return word_dict
````

* El parámetro `word` es una lista de cadenas (`list[str]`). 
* El valor de retorno es un  diccionario (dict) donde:
    * Las claves (`str`) son palabras de la lista word.
    * Los valores (`list[int]`) son listas de enteros que representan las posiciones de inicio donde cada palabra aparece en el texto.

````python
texto = "hello world, hello universe"
palabras = ["hello", "world", "universe"]

resultado = find_word(texto, palabras)
print(resultado)
````

## Tipos genéricos

Los tipos genéricos permiten definir estructuras de datos o funciones que pueden trabajar con distintos tipos manteniendo la información de tipos.

````python
from typing import TypeVar

T = TypeVar("T")

def primero(elementos: list[T]) -> T:
    return elementos[0]

resultado = primero([100, 'Hola', 4.5])
````

T pasa a ser la unión de todos los tipos presentes en la lista:

```python
    resultado: int | str | float
```

## Unión de tipos

La función `funcion_ejemplo` siguiente está diseñada para recibir un parámetro `x` que puede ser un número entero (`int`) o `None` y realiza una acción dependiendo de si `x` es `None` o no. La función no tiene un valor de retorno.

````python
def funcion_ejemplo(x: int | None) -> None:
    if x is not None:
        print(x)
    else:
        print('El parámetro es None')
````

## IO

Para funciones que aceptan o devuelve  objetos provenientes de una llamada a `open()`  se usa el tipo `IO[str]` (para archivos de texto) o `IO[bytes]` (para archivos binarios). IO no distingue entre modos de lectura, escritura u otros.


````python
from typing import IO

def escribir_en_archivo(archivo: IO[str], contenido: str) -> None:
    archivo.write(contenido)
    
# Uso con un archivo de texto
with open("archivo.txt", "w", encoding="utf-8") as f:
    escribir_en_archivo(f, "Hola, mundo!")
````

## Alias de tipos

Esto es nuevo  en la versión 3.12 de Python. Permiten asignar un nombre descriptivo a un tipo complejo para mejorar la legibilidad y facilitar el mantenimiento del código. 



````python
type Numero = int | float

def cuadrado(x: Numero) -> Numero:
    return x * x

cuadrado(8)
````

````python
type IntNone = int | None   
def funcion_ejemplo(x: IntNone ) -> None :
    if x is not None :
        print (x)
    else :
        print ('El parametro es None')
````

Resulta especialmente útil cuando se trabaja con tipos complejos, como uniones o estructuras de datos anidadas.


````python
type WordIndex = dict [str , list[int]]

def find_word( text : str , word : list[str]) -> WordIndex:
    word_dict = {}
    for w in word :
        word_dict[w] = []
    i = 0
    while i < len( text ):
        for w in word :
            if text [i:i+ len(w)] == w:
                word_dict[w].append(i)
        i += 1
    return word_dict
````

También es posible definir alias genéricos:

````python
type Lista[T] = list[T]

numeros: Lista[int] = [1, 2, 3]
nombres: Lista[str] = ["Ana", "Luis"]
````

## Funciones

Callable es un tipo genérico que representa cualquier objeto invocable, es decir, cualquier objeto que pueda llamarse como una función.

````python
from typing import Callable

def aplicar( funcion: Callable[[int, int], int],
                   x: int,
                   y: int ) -> int:
    return funcion(x, y)

def multiplicar(a: int, b: int) -> int:
    return a * b

resultado = aplicar(multiplicar, 3, 4)
resultado
````

Sin especificar los parámetros:

````python
from typing import Any, Callable

funcion: Callable[..., Any]
````

El ... significa cualquier número de parámetros de cualquier tipo.

````python

def saludar(nombre: str) -> str:
    return f"Hola {nombre}"

f: Callable[..., object] = saludar
````

### Desventajas del uso de anotaciones de tipo

Hay tres posibles desventajas de usar anotaciones de tipo:

* El código con anotaciones de tipo requiere  más tiempo para escribir  código. Por una parte implica literalmente más escritura (teclear más en el teclado) y requiere razonar más sobre el código. Razonar sobre el código casi siempre es algo positivo. Sin embargo, es de esperar que ese tiempo perdido se recupere al detectar errores con un verificador de tipos antes de que el programa se ejecute.

* Las anotaciones de tipo, en algunos casos, pueden disminuir la legibilidad. Esto puede deberse a dos razones: la falta de familiaridad y la verbosidad. Cualquier sintaxis que no  resulte familiar será menos legible que una sintaxis conocida. Las anotaciones de tipo realmente cambian la apariencia de los programas en Python, haciéndolos potencialmente desconocidos al principio. Esto solo puede resolverse escribiendo y leyendo más código Python con anotaciones de tipo. El compromiso es que la comprensión del código puede ser mayor tras una primera lectura, incluso si esa lectura lleva más tiempo. Con las anotaciones de tipo, se ven  inmediatamente todos los tipos esperados, lo cual es una ventaja en comparación con tener que examinar el código en sí para entender los tipos o leer la documentación.



Las anotaciones de tipo todavía están en evolución. Han mejorado significativamente desde que fueron introducidas por primera vez en Python 3.5, pero aún hay casos particulares donde no funcionan bien. Aun así, proporcionan muchos beneficios, así que podemos utilizar las anotaciones de tipo de manera parcial.
