# Herramientas necesarias I


Para poder ejecutar los ejemplos es necesario tener accesible un intérprete de Python 3.X. La versión usada en este curso es la versión Python 3.14, pero podemos usar cualquiera de las anteriores (3.9, 3.11, 3.12, 3.13). 

El entorno de desarrollo que vamos a usar aquí es __PyCharm__ y puede ser buena idea que usemos todos el mismo, pero por supuesto no es obligatorio. Otros entornos de desarrollo proporcionan características similares a las de PyCharm.

__PyCharm__ es un entorno de desarrollo integrado (IDE) especializado para el desarrollo en Python. Desarrollado por la compañía __JetBrains__, PyCharm es muy popular entre los programadores debido a que proporciona un gran número de  herramientas que facilitan el trabajo en proyectos complejos y en una variedad de marcos y bibliotecas en Python, como Django, Flask, Pandas y PySpark entre otros.

Entre las herramientas integradas en PyCharm encontramos:

* Editor de código con **autocompletado** avanzado.
* **Refactorización** de código. Preguntarme si no conocéis esta utilidad (es maravillosa).
* **Gestión avanzada de proyectos** con múltiples archivos, carpetas, y dependencias. Para gestionar proyectos, Jupyter no es lo más apropiado ya que  trabaja con notebooks individuales que, aunque son útiles para experimentos, pueden volverse desordenados en proyectos grandes.
* **Depuración** paso a paso. Podremos establecer puntos de interrupción (breakpoints), inspeccionar variables en tiempo de ejecución, así como hacer seguimiento detallado del flujo del programa.
* PyCharm tiene **integración nativa con Git** y otros sistemas de control de versiones.

# Instalación de Pycharm

Hay 2 versiones de PyCharm:
* `Community` Libre y Gratuito, pero versión con funcionalidad limitada.
* `Professional` <p style="color:blue;">De pago por suscripción. Los alumnos UCM pueden registrarse con licencia educativa (gratis por un año y renovable). Para ello, registrarse con correo UCM. </p>

La instalación de PyCharm depende del sistema. Se descarga el instalador de la página https://www.jetbrains.com/pycharm/ y se sigen las instrucciones.

En Ubuntu/Linux, sin embargo, la forma más sencilla es usar el repositorio snap. Basta abrir un terminal y ejecutar el comando
```
$ sudo snap install pycharm - professional
```

## Características

<img src="images/py_main_window_overview.png" alt="Anaconda instaler" style="width: 500px; float:center"/>

1. La __cabecera de la ventana__ contiene un conjunto de widgets que proporcionan acceso rápido a las acciones más populares: el widget del proyecto y el widget de ejecución. También permite la opción de "Buscar" y las opciones de configuración.

2. La __ventana del proyecto__, ubicada en el lado izquierdo, muestra los archivos del proyecto.

3. El __editor__, en el lado derecho, es donde realmente podremos escribir el código. Cuenta con pestañas para facilitar la navegación entre los archivos abiertos.

4. Los __menús contextuales__ se abren al hacer clic derecho sobre un elemento o un fragmento de código, y muestran las acciones disponibles.

5. La __barra de navegación__ permite navegar rápidamente por las carpetas y archivos del proyecto.

6. La __franja vertical__ junto al editor, muestra los puntos de interrupción (breakpoints) que tiene un script de python y proporciona una forma conveniente de navegar por la jerarquía del código, como ir a definiciones o declaraciones. También muestra los números de línea.

7. La barra de desplazamiento, ubicada en el lado derecho del editor.

8. Las ventanas de herramientas son ventanas especializadas adjuntas en la parte inferior y en los laterales del espacio de trabajo. Proporcionan acceso a tareas típicas como la gestión del proyecto, búsqueda y navegación en el código fuente, integración con sistemas de control de versiones, ejecución, pruebas, depuración, y más.

9. La __barra de estado__ indica el estado de tu proyecto y de todo el IDE, mostrando varios mensajes de advertencia e información como codificación de archivos, separador de líneas, perfil de inspección, y más. También ofrece acceso rápido a la configuración del intérprete de Python.



En el lado izquierdo de la ventana de la aplicación nos encontramos un menú de herramientas (Tool Window Bar) desde el cual podremos abrir la consola de python, el gestor de paquetes, la terminal, etc.

## Entornos virtuales

Más adelante hablaremos más de los entornos virtuales. Aquí damos algunas pinceladas en su relación con el entorno PyCharm.

Los entornos virtuales permiten separar las versiones de python entre distintos proyectos en los que estemos trabajando o de la
versión de python del sistema. 

__Ventajas__:

* Permiten **cambiar la versión de python fácilmente entre proyectos**. Python es un lenguaje en evolución y es posible que tengamos que garantizar que el proyecto sea compatible con una versión anterior de python a la que tengamos nosotros en el sistema.

* **Cada entorno virtual puede tener sus propios paquetes**. Es necesario recordar, que los paquetes instalados en un entorno virtual no están necesariamente en otro entorno virtual.

En PyCharm el entorno virtual se elige al principio. La elección inical se puede cambiar en las preferencias.



Como se puede ver, podremos seleccionar un entorno virtual ya creado o crear un entorno virtual nuevo (con el nombre que queramos y la versión de Python más apropiada para nuestro proyecto.

## Instalación de paquetes en nuestro entorno virtual

El servicio Pypi https://pypi.org/ pone a disposición más de 480000 paquetes como Pandas, jupyter.
La instalación estándar es usar un terminal y ejecutar el comando `pip` con una lista de paquetes.

```
$ pip install pandas jupyter 
```

Para que se puedan usar en nuestro proyecto es conveniente instalarlos desde la terminal de PyCharm.

## Cuadernos de Jupyter

Son archivos con extensión .ipynb. En ellos podemos escribir código python ejecutable, texto, dibujar gráficas y mucho más.

PyCharm soporta los cuadernos de jupyter para ello es necesario instalar el paquets `jupyter`. Desde el terminal
```
 pip install jupyter 
```
Luego se puede integrar en el interfaz de PyCharm desde el menú Tools.



También se pueden abrir desde el terminal con el comando:
```
jupyter notebook
```



La ejecución de `jupyter notebook` abrirá una ventana nueva en el navegador de internet.
