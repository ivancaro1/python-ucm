# Intérprete, dependencias y entornos virtuales

En este notebook aprenderemos a distinguir las piezas que intervienen cuando ejecutamos un programa Python y a crear **entornos reproducibles**. 


---
## 1. Conceptos

### Editor o IDE 

Un **IDE** (Entorno de Desarrollo Integrado) es una aplicación que facilita la programación al reunir en una sola interfaz herramientas como el editor de código, la ejecución de programas, la depuración de errores y otras utilidades. 

En Python existen numerosos IDEs:
* **IDLE** (incluido con Python)
* **PyCharm**
* **Visual Studio Code** 
* **Spyder**

Todos ellos ayudan a escribir, ejecutar y organizar proyectos de forma cómoda y productiva.



### El intérprete de Python

Python es un lenguaje interpretado, siendo  el **intérprete  el programa que lee y ejecuta código Python**.

Cuando instalamos Python, en nuestro equipo tendremos el ejecutable `python.exe` (en Windows), que es el interprete de Python. 

* Sí, en un equipo podemos instalar varios intérpretes, es decir, distintas versiones de Python.

El intérprete de Python puede ejecutarse desde la consola mediante el comando _python_ (o _python3_, según el sistema operativo), siempre que la instalación haya añadido al PATH el directorio donde se encuentra el ejecutable de Python.

__Biblioteca estándar de Python__

Dentro de la instalación de Python existen numerosas carpetas y archivos que permiten que el lenguaje funcione correctamente. Una de las más importantes es la carpeta **Lib**, ya que contiene la **biblioteca estándar de Python**, un conjunto de módulos que vienen incluidos con Python y que podemos utilizar directamente en nuestros programas. 

Por ejemplo, el módulo `math` permite realizar operaciones matemáticas avanzadas de forma sencilla. 

__Gestor de paquetes__

Si queremos ampliar las funcionalidades de Python, podemos instalar nuevas bibliotecas o paquetes. Lo habitual es utilizar `pip`, el gestor de paquetes de Python, que descarga e instala automáticamente tanto el paquete deseado como sus dependencias. La mayoría de estos paquetes se encuentran en **PyPI** (https://pypi.org/), el repositorio oficial de paquetes de Python. De este modo, `pip` facilita la gestión de las bibliotecas disponibles en nuestra instalación de Python.

Cuando instalamos nuevas bibliotecas, éstas suelen almacenarse en una carpeta llamada **site-packages**, que forma parte de la instalación de Python (la estructura exacta de carpetas puede variar entre Windows, Linux y macOS).


__Entorno global (base)__

Para una instalación concreta, por ejemplo **Python 3.12**, existe un único espacio global asociado llamado `entorno global (base)` entorno global podría ser:

```powershell
    Python 3.12
    ├── Biblioteca estándar
    └── Site-packages (paquetes instalados de forma global)
        └──pip
```

Podemos tener varios entornos globales, uno por cada instalación de Python. Por ejemplo, podemos tener instaladas simultáneamente: **Python 3.10**, **Python 3.12** y **Python 3.13** en el mismo equipo, cada una de ellas con su versión de pip y sus paquetes.

```powershell
    ├── Python 3.10 (base)  
    │   ├── Biblioteca estándar
    │   └── Site-packages
    │        └──pip
    |
    ├── Python 3.12 (base)
    │   ├── Biblioteca estándar
    │   └── Site-packages
    │        └──pip
    |
    └── Python 3.13 (base)
        ├── Biblioteca estándar
        └── Site-packages
            └──pip
    
```


````python
# En Windows podemos ver los intérpretes disponibles con:
!py -0p
````

````python
# Para ejecutar un programa con Python 3.13:
!py -3.13 ./main.py
````

Los entornos globales no se “activan” , pero sí se seleccionan usando ```py -3.11```, ```py -3.13```.

### Notebook y Kernel 

Un **notebook** es documento con texto, código Python y resultados.

El **kernel** es una sesión concreta que utiliza un intérprete determinado para ejecutar las celdas del notebook. Se puede decir que es un proceso que ejecuta las celdas de código y mantiene el estado de un notebook en memoria.

Si reiniciamos  el kernel:

* elimina las variables de memoria
* borra los imports realizados
* detiene ejecuciones en curso
* permite empezar desde un estado limpio

__Kernel e intérprete incorrectos__: Un problema muy frecuente es instalar un paquete en un entorno y ejecutar el notebook con un kernel que use otro entorno.

* Normalmente, no existe el concepto de kernel en un archivo `.py`. Un archivo .py se ejecuta directamente con un intérprete:

    archivo.py → intérprete de Python → resultado


En `VS Code` debemos seleccionar el kernel desde la parte superior derecha del notebook y elegir el intérprete asociado al entorno virtual correcto.

````python
import sys

# Indica la ruta del ejecutable de Python que está usando el proceso actual. 
# Sirve para comprobar si el notebook está usando el entorno correcto.
print('Ejecutable:', sys.executable)
# Indica la versión y los detalles de compilación del intérprete.
print('Versión:', sys.version)
````

La ruta mostrada por `sys.executable` es una de las comprobaciones más útiles. Si el terminal dice que un paquete está instalado pero el notebook no lo encuentra, normalmente estamos usando dos intérpretes diferentes.

### Módulos, paquetes y dependencias

* Un **módulo** es normalmente un archivo `.py` que contiene código reutilizable. 
* Un **paquete** es una distribución de código que puede incluir uno o varios módulos, metadatos y otros recursos.

<p align="center"><img src="images/modulos_n1.png" width="600"></p>

* Una **dependencia** es algo que nuestro proyecto necesita para funcionar: otro paquete, una versión concreta de Python o incluso una herramienta del sistema. 

    Por ejemplo, un proyecto que importa `pandas` tiene `pandas` como dependencia directa; las librerías que `pandas` necesita (por ejemplo `numpy`) son dependencias transitivas.

Un problema frecuente es que **distintos proyectos requieren versiones diferentes de las mismas librerías**.

**Ejemplo**:

Dos proyectos que usen pandas con distintas versiones. Si ambas versiones se instalan globalmente pueden surgir conflictos. Para evitarlo se crean **entornos virtuales** (hablaremos más adelante):

<p align="center"><img src="images/envs.png"  width="600"></p>

## 2. Sistemas de gestión de paquetes

Son herramientas que automatizan la instalación y administración de software o bibliotecas. `pip` es el gestor de paquetes oficial de Python.

Estos sistemas permiten:

* Instalar paquetes
* Resolver dependencias
* Actualizar versiones
* Desinstalar paquetes
* Mantener un registro de lo instalado (mediante **requirements.txt**)

```python

    # pip instala pandas y también las bibliotecas que necesita para funcionar.
    pip install pandas
    

```

Los gestores de paquetes interactúan con un repositorio donde se almacenan los paquetes. En el caso de pip, el repositorio oficial es [PyPI](https://pypi.org/). 
`pip` busca el paquete en PyPI, descarga el archivo correspondiente y lo ubica en la carpeta de librerías de tu entorno (site-packages).

Un paquete publicado en PyPI puede distribuirse en varios formatos:

- **wheel** (`.whl`): archivo preparado para instalarse, parecido a un ZIP, con código compilado, metadatos e información sobre dependencias. Por ejemplo:

```powershell

    pandas-2.2.3-cp311-cp311-win_amd64.whl
```

- **distribución de código fuente** (`sdist`): Contiene el código fuente y los archivos necesarios para construir e instalar el paquete. En algunos casos, el equipo debe compilar parte del paquete, por lo que pueden ser necesarios compiladores u otras herramientas.


- **metadatos**: nombre, versión, dependencias, licencia y compatibilidad.

`pip` busca una distribución compatible con nuestro sistema y nuestro intérprete. El nombre que usamos para instalar puede no coincidir exactamente con el nombre que importamos en Python.

Otros gestores de paquetes son `uv`, `conda`.

## 3. Muy interesante ... Los entornos virtuales

* Un entorno virtual es una instalación aislada de paquetes asociada a un intérprete base.

* Los entornos virtuales **funcionan de forma independiente**, por lo que cada uno puede tener su propia versión de Python y sus propias bibliotecas, aunque coincidan con las de otros entornos. 

* Su principal ventaja es __aislar las dependencias de cada proyecto__, permitiendo que varios proyectos con requisitos distintos convivan en la misma máquina sin interferirse entre sí.

    > Aunque suele decirse que un entorno virtual es una copia del entorno global o base, en realidad **no se duplican todos los archivos**. Para ahorrar espacio y acelerar su creación, algunos componentes se comparten o se referencian desde la instalación base. Por ello, los cambios importantes realizados en el entorno global pueden afectar a los entornos virtuales que dependan de él.

* **Buenas prácticas**: Aunque un entorno virtual puede compartirse entre varios proyectos, lo más recomendable es asignar uno exclusivo a cada proyecto para garantizar un aislamiento total.

### Gestores de paquetes y relación con los entornos

Los gestores de paquetes instalan paquetes en un entorno determinado:

```powershell
    Python 3.12  (base)
    │
    ├── Intérprete
    ├── Biblioteca estándar
    ├── Site-packages 
    │   └── pip
    │
    └── Entornos virtuales
        ├── venv1
        │   └── Site-packages 
        |       └── pip
        └── venv2
            └── Site-packages 
                └── pip
```

* Si el entorno activo es el base, los paquetes se instalan allí.

* Si el entorno activo es un entorno virtual, los paquetes se instalan únicamente en ese entorno.


### Ubicación de los entornos 

Los entornos virtuales son simplemente carpetas que contienen una instalación aislada de Python y sus dependencias. **Su ubicación la decide el desarrollador al crearlos**.

__En la carpeta del proyecto__.

Opción práctica para proyectos pequeños o personales. Se tiene un control total del entorno.

```powershell
        mi_proyecto/
        ├── programa.py
        └── .venv/
            ├── Include/
            ├── Lib/
            │   └── site-packages/
            ├── Scripts/
            └── pyvenv.cfg
        └── requirements.txt
```

La carpeta .venv contiene:

* el acceso al intérprete de Python
* los paquetes instalados para ese entorno
* la configuración que indica con qué intérprete de Python se creó
* los scripts de **activación del entorno**

**En una ubicacición centralizada**

Opción más adecuada si el proyecto se va a compartir con diferentes máquinas.

```cmd
        C:\Users\adolfo\venvs\
        ├── proyectoA
        ├── proyectoB
        └── proyectoC
```

Esta opción evita errores de rutas absolutas (rutas fijas a la instalación de Python) y evita compartir carpetas grandes de librerías instaladas. Solo será necesario compartir  un archivo de texto con la lista de librerías necesarias para poder generar un entorno compatible en cada máquina (como `requirements.txt` o `pyproject.toml` de los que hablaremos más adelante) .

### Crear y usar entornos virtuales con `venv`

El flujo de trabajo básico para gestionar entornos virtuales en Python utilizando el módulo nativo `venv` comprende la creación, activación, uso y desactivación del entorno.

Desde la carpeta del proyecto:

| Acción | Windows (PowerShell / CMD)  | |
| :--- | :--- | :--- |
| **Crear entorno** | `python -m venv mi_entorno`   | Esto generará una carpeta llamada mi_entorno (o .venv) con una copia aislada del intérprete|
| **Activar (PowerShell)** | `.\mi_entorno\Scripts\Activate.ps1` ||
| **Activar (CMD)** | `.\mi_entorno\Scripts\activate.bat`  | Al activar el entorno, la terminal mostrará el nombre entre paréntesis (mi_entorno) al inicio de la línea de comandos. A partir de este momento, cualquier paquete instalado afectará solo a este entorno |
| **Instalar paquetes** |`pip install nombre_paquete`||
| **Desactivar** | `deactivate` | |


**Buenas prácticas**: Si trabajamos con Git, es recomendable añadir el nombre de la carpeta de tu entorno virtual (por ejemplo, mi_entorno/, venv/ o .venv/ ) al archivo .gitignore para no subir los archivos binarios del entorno al control de versiones.

### Fichero de requisitos `requirements.txt`

Un fichero de requisitos es una lista sencilla de dependencias asociado a un entorno virtual.

```text
    # requirements.txt
    requests>=2.31,<3
    pandas~=2.2
    pytest>=8
```
* Desde la terminal podemos generar el archivo `requiements.txt` 

```python
        pip freeze > requirements.txt
```
* Una vez creado puede usarse para reconstruir un entoirno en otra máquina

```python
        pip install -r requirements.txt
```

En ambos casos, es necesario tener el entorno virtual activo (si no lo está ya). Para activar entorno, desde la termianl ejecutar:
```python
        .\mi_entorno\Scripts\activate.bat
```   
Al activar el entorno, la terminal mostrará el nombre entre paréntesis (mi_entorno) al inicio de la línea de comandos. A partir de este momento, cualquier paquete instalado afectará solo a este entorno.

Para controlar qué versión de un paquete se instala y evitar incompatibilidades con el código, se utilizan **operadores de versión**. Los especificadores se escriben normalmente con la sintaxis de la especificación de dependencias de Python.

Ejemplos:

| Especificador | Significado aproximado |
|---|---|
| `pandas==2.2.3` | Instala únicamente la versión 2.2.0.|
| `pandas>=2.0` |  Instala la 2.0.0 o cualquiera superior |
| `pandas<3` | Evita instalar versiones mayores a la especificada. |
| `pandas>=2,<3` | Rango cerrado. Instala cualquier versión entre la 2 y la 3 (excluyendo la 3). |
| `pandas!=2.1.0` | Excluye una versión 2.1.0 |
| `pandas ~= 2.1.0` |(Operador de compatibilidad): Equivalente a >= 2.1.0, == 2.1.*. Permite actualizaciones de parches dentro de la misma versión menor, pero evita saltar a la 2.2.0. |

`requirements.txt` es práctico para describir instalaciones, pero por sí solo no siempre registra toda la resolución transitiva de forma reproducible.

## 4. El gestor moderno de proyectos Python `uv` 

`uv` es una **herramienta de sistema** de línea de comandos (CLI tool) diseñada para administrar el propio ecosistema de Python (instalar paquetes, crear entornos virtuales y gestionar versiones del lenguaje).

Entre las características más importantes:

* Ejecuta instalaciones de librerías entre 10 y 100 veces más rápido que `pip` gracias a su paralelización y un sistema de caché global muy eficiente.

* Integra tareas que tradicionalmente hacemos con otras herramientas (`venv`, `pip`, `pip-tools` y gestores de herramientas).

La relación con las herramientas clásicas puede resumirse así:

- `venv` crea un entorno virtual
- `pip` instala paquetes en un entorno
- `uv` reúne y automatiza gran parte de ese trabajo:
    * gestiona eficazmente las dependencias de los paquetes y los entornos virtuales
    * proporciona herramientas para inicializar y estructurar proyectos Python 
    * genera archivos de bloqueo para garantizar entornos reproducibles (`uv.lock`, `pyproject.toml`)
    * soporta la publicación de paquetes en PyPI
    * se integra perfectamente con los estándares y herramientas de empaquetado de Python existentes

### Instalación en Windows

En https://docs.astral.sh/uv/getting-started/installation/#__tabbed_1_1 podemos encontrar instrucciones para la instalación de `uv` dependiendo del sistema operativo (se admite la instalación con PIP, pero no se recomienda).

Una vez instalado, escribir en consola:

```powershell
   uv --version
```



### Crear un proyecto

Para iniciar un proyecto vacío:

```powershell
    uv init nombre_proyecto
    cd nombre_proyecto
```

Esto crea una carpeta llamada nombre_proyecto con el siguiente contenido:

```powershell
    nombre_proyecto
    ├── .gitignore
    ├── .python-version
    ├── README.md
    ├── main.py
    └── pyproject.toml
```

Si queremos preparar un proyecto que se pueda instalar como paquete:

```powershell
    uv init --package example-pkg
```

Esto creará una carpeta con distintos archivos y carpetas:

```powershell
    example-pkg
    ├── src
    ├── .gitignore
    ├── .python-version
    ├── README.md
    └── pyproject.toml
```

El archivo `.python-version` contiene la versión de Python utilizada para el proyecto, y será la que se use posteriormente para crear el entorno virtual. Por otro lado,  `pyproject.toml` sirve como archivo de configuración principal para los metadatos y dependencias del proyecto. 


### Añadir y eliminar dependencias

* `uv` usa el comando `add` para la instalación de dependencias. 
* La primera vez que se  ejecuta el comando `add`, se crea un nuevo entorno virtual en la carpeta del proyecto e instala las dependencias especificadas. 
* Al igual que `pip`, `uv`, normalmente descarga desde PyPI.

```powershell
    cd nombre_proyecto
    uv add numpy scikit-learn
    uv remove numpy scikit-learn
```

Al ejecutar `uv add <paquete>`, se realiza cuatro acciones automáticamente:

1. Resuelve la versión óptima compatible con tu proyecto.
2. Escribe la dependencia en `pyproject.toml`.
3. Actualiza el árbol de dependencias y hashes en `uv.lock`.
4. Instala el paquete en tu `.venv`.

El entorno suele encontrarse en la carpeta `.venv`.

```powershell
    nombre_proyecto
    ├── .venv
        ├── Lib
        └── Scripts
    ├── .gitignore
    ├── .python-version
    ├── README.md
    ├── main.py
    ├── uv.lock
    └── pyproject.toml
```

El aspecto de `pyproject.toml` :

```powershell
    [project]
    name = "nombre-proyecto"
    version = "0.1.0"
    description = "Add your description here"
    readme = "README.md"
    requires-python = ">=3.14"
    dependencies = [
        "numpy>=2.5.3",
    ]
```

Mientras que `.python-version` afecta al entorno de desarrollo, `pyproject.toml` establece las reglas y características del proyecto para todos los miembros del equipo.




Los archivos `pyproject.toml` y `uv.lock` son los que se encargan de la gestión de dependencias. Cumplen roles complementarios para equilibrar la flexibilidad y la reproducibilidad.

| Característica | pyproject.toml | uv.lock |
| :--- | :--- | :--- |
| **Generado por** | El desarrollador (o `uv add`) | `uv` automáticamente |
| **Edición manual** | Sí | No |
| **Contenido** | Dependencias directas requeridas | Almacena el grafo de todo el árbol de dependencias + el hash SHA-256 del archivo descargado desde PyPI  |
| **Nivel de versión** | Flexible (`>=`, `~=`) | Exacta (`==`) |
| **Propósito** | Definir qué necesita el proyecto | Garantizar instalaciones idénticas |
| **Se comparte en Git** | Sí | Sí |


Por otro lado `uv remove` realiza automáticamente tres acciones en un solo paso:

1. Quita la dependencia del archivo `pyproject.toml`.
2. Revisa y actualiza el archivo de bloqueo `uv.lock` (eliminando también las sub-dependencias que ya no sean necesarias).
3. Elimina el paquete del entorno virtual (`.venv`).

### Limpieza de paquetes no usados

El comando `uv sync` compara el estado de tu entorno virtual `.venv` contra el archivo `uv.lock`. 
Si encuentra librerías dentro del entorno que no figuran en el archivo de bloqueo, las detecta como paquetes no usados y las elimina automáticamente para garantizar un entorno limpio y exacto.

```powershell
    uv sync --prune
```


### Ejecutar código del proyecto

```powershell
    uv run python main.py
```

`uv run` ejecuta el comando usando el entorno y las dependencias declaradas por el proyecto. Si es necesario, sincroniza primero el entorno.


### Más operaciones con uv

#### Versiones de Python

```cmd
uv python list --only-installed
```

La versión compatible con el proyecto se declara en `pyproject.toml`:

```cmd
requires-python = ">=3.11,<3.14"
```


Para cambiar la versión de Python en un proyecto gestionado con `uv`, primero se instala la versión deseada (si no está disponible en el sistema):

```cmd
 uv python install 3.14
 ```

 después se fija como versión del proyecto:
 
```cmd
uv python pin 3.14
```

lo que actualiza el archivo `.python-version`.

Finalmente, es necesario sincronizar el entorno virtual con la nueva versión de Python.

```cmd
uv sync
´´´

Actualizar la restricción en pyproject.toml (requires-python = ">=3.14")

#### Ejecutar herramientas sin instalarlas en el proyecto

Para ejecutar herramientas como `mypy`, `black` o `pytest` sin añadirlas como dependencias del proyecto podemos usar:

```cmd
uv tool run mypy hello.py
uvx mypy hello.py
```

`uvx` es una forma abreviada de `uv tool run`. `uv` crea o reutiliza un entorno temporal en su caché, instala allí la herramienta y la ejecuta. Así no se modifica el entorno global ni se ensucia el entorno del proyecto.

Para ejecutar una herramienta que sí forma parte del proyecto, usamos normalmente:

```cmd
    uv run pytest
    uv run black 
```


#### Exportar dependencias

Podemos exportar las dependencias resueltas a un fichero compatible con `pip`:

```cmd
    uv export -o requirements.txt
```

#### Migrar desde `pip`

Una migración sencilla puede ser:

```cmd
    pip freeze > requirements.txt
    uv init .
    uv pip install -r requirements.txt
```

Es preferible revisar después `pyproject.toml` para declarar las dependencias directas del proyecto, en lugar de conservar únicamente toda la lista congelada.

#### Kernels de Jupyter 

Para consultar los kernels registrados:

```cmd
jupyter kernelspec list
```

El kernel seleccionado debe corresponder al intérprete del entorno `.venv` del proyecto. Podemos comprobarlo desde una celda Python con:


````python
import sys

print(sys.executable)
print(sys.version)
````

Para que Jupyter Notebook o JupyterLab detecten y utilicen el entorno virtual generado por uv, debes registrar dicho entorno como un kernel de Jupyter instalando la librería `ipykernel`.

1. Añade `ipykernel` a las dependencias de tu proyecto. `uv` se encargará de instalarlo dentro del entorno virtual .venv.

```cmd
    uv add ipykernel
```

2. Registrar el entorno como Kernel en Jupyter

```cmd
    uv run python -m ipykernel install --user --name=mi-proyecto --display-name "Python (mi-proyecto)"
```
__Cómo usarlo en Jupyter__

* Desde VS Code / PyCharm: Abre tu cuaderno .ipynb, haz clic en la esquina superior derecha (Select Kernel / Seleccionar Kernel) y elige la opción que corresponde al entorno virtual de tu proyecto o el nombre asignado en --display-name.

* Desde la terminal (JupyterLab): Si ejecutas JupyterLab , el entorno estará disponible automáticamente como opción predeterminada.

```cmd
  uv run jupyter lab
```
