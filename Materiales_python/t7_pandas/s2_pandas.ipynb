### 3.1.1 Acceso a columnas




Para acceder a una columna usamos corchetes con el nombre de la columna. El resultado es una Serie:

### 3.1.2 Seleccionar filas por posición: `.iloc[]`

`.iloc[]` selecciona filas (y columnas) por su **posición numérica** (empezando por 0):

````python
# Primera fila
hoteles.iloc[0]
````

````python
# Filas 2 a 4 (sin incluir la 5)
hoteles.iloc[2:5]
````

### 3.1.3 Seleccionar filas por etiqueta: `.loc[]`

`.loc[]` selecciona filas por su **etiqueta de índice** y columnas por su **nombre**.

Para ver la diferencia con `.iloc[]`, vamos a crear un DataFrame donde el índice **no** sea numérico. Cargaremos los datos de hoteles usando el codigo como índice:

````python
hoteles2 = pd.read_csv("datos/hoteles.csv", 
                      sep = ';', 
                      parse_dates = ['date'], 
                      date_format='%d/%m/%Y',
                      index_col = ['code'])

# Veamos las primeras filas
hoteles2.head()
````

````python
# .loc[] usa las ETIQUETAS del índice (los códigos de hoteles)
print("--- .loc[] selecciona por etiqueta ---")
hoteles2.loc["AA56"]
````

````python
# Rango de etiquetas (SÍ incluye el final, a diferencia de .iloc[])
hoteles2.loc["TH46":"AA56"]
````

````python
# Seleccionar filas y columnas a la vez con .loc[]
hoteles2.loc["TH46":"AA56", ["nombre", "precio"]]
````

Ahora la diferencia entre `.iloc[]` y `.loc[]` queda clara:
- **`.iloc[0]`** → la primera fila (posición 0), sea cual sea su etiqueta
- **`.loc["TH46"]`** → la fila con etiqueta `"TH46"`, esté en la posición que esté

### 3.1.4 Resumen: `.iloc[]` vs `.loc[]`

| Método | Selecciona por | Usa | Rango incluye final | Ejemplo |
|--------|---------------|-----|--------------------|---------|
| `.iloc[]` | **Posición** | Números enteros | No | `df.iloc[0:3]` → posiciones 0, 1, 2 |
| `.loc[]` | **Etiqueta** | Nombres/etiquetas del índice | Sí | `df.loc["POL-001":"POL-003"]` → POL-001, POL-002, POL-003 |

## 3.2 Filtrado de datos

El filtrado es una de las operaciones más frecuentes. Permite seleccionar solo las filas que cumplen una condición.

### 3.2.1 Filtrar con una condición

````python
# hoteles con precio > 100
# 1. Creamos a condición
condicion = hoteles["precio"] > 100
# 2. Aplicamos la condición con los corchetes
hoteles100 = hoteles[condicion]

hoteles100.head()
````

````python
len(hoteles), len(hoteles100)
````

````python
# hoteles ubicados en el distrito centro
# 1. Concicion
condicion2 = hoteles["distrito"] == "Centro"
# 2. Aplicamos la condición con los corchetes
hoteles_centro = hoteles[condicion2]

hoteles_centro.head()
````

### 3.2.2 Combinar condiciones

Podemos combinar varias condiciones usando:
- `&` para **Y** (and): ambas condiciones deben cumplirse
- `|` para **O** (or): al menos una debe cumplirse

**Importante:** cada condición debe ir entre paréntesis.

````python
# hoteles distroto Centro o Retiro

# 1. Conciciones 
filtro_centro = hoteles["distrito"] == "Centro"
filtro_retiro = hoteles["distrito"] == "Retiro"
# 2. Aplicamos la condición con los corchetes
hotelesCentroRetiro = hoteles[filtro_centro | filtro_retiro]
hotelesCentroRetiro.head()
````

**Importante:** cada condición debe ir entre paréntesis.

````python
# hoteles distro Centro con precio superior a 80
hotelesCentro100 = hoteles[(hoteles["distrito"] == "Centro") & (hoteles["precio"] > 80)]
hotelesCentro100.head()
````

### 3.2.3 Filtrar con `.isin()`

Cuando queremos filtrar por varios valores de una misma columna, `.isin()` es más cómodo que encadenar condiciones con `|`:

````python
#hoteles distroto Centro o Retiro
distritos = ["Centro", "Retiro"]
hotelesCR = hoteles[hoteles["distrito"].isin(distritos)]
hotelesCR.head()
````

## 4.1 Missing Values (NaN)

Los datos o valores perdidos (missing data) están presentes en la mayorías de los conjuntos de datos. Son aquellos en los que no se almacena ningún valor. Una de las principales tareas para el análisis de datos suele ser la evaluación de la existencia de datos perdidos.

* La ausencia de valores en los conjuntos de datos puede tener varios efectos perjudiciales:
  
    - Limitan la capacidad para realizar tareas de ciencia de datos, como la conversión de tipos de datos o la visualización de datos.
    - En general, reducen la validez estadística de los modelos de ML.
    - Nos pueden llevar a conclusiones no válidas.

En Python, estos valores se identifican con `NaN`(Not a Number). Aunque Pandas es capaz de tratar este tipo de valores de forma implícita, resulta de mucha utilidad conocer las herramientas que proporciona Pandas para su manejo.

### 4.1.1 Identificacion de nulos en el dataframe

````python
# Devuelve True si el valor es distinto de  nulo
hoteles.notna()    # hoteles.notnull()
````

````python
# cálculo de nulos por columnas
hoteles.isna().sum()
````

### 4.1.3 Imputación de valores nulos

En los casos en los no deseamos eliminar datos con valores `NaN`, Pandas ofrece la posibilidad de sustituir dichos  valores por un valor por defecto. 
Para ello tenemos varias opciones:

* el método `fillna()` junto con el valor a sustituir.
* método `ffill()` para propagar el ultimo valor válido hasta el siguiente valor válido. (Propagación hacia adelante)
* método `bfill()` sustituye los NaN por el primer valor válido. (Propagación hacia atrás)

#### 4.1.3.1 Reemplazo por valor fijo

````python
# sustituye todos los NaN por 0
hoteles.fillna(0)
````

También es posible indicar un valor por defecto distinto para cada columna. En ese caso,  usaremos un diccionario como argumento del método `fillna`. Las claves del diccionario serán las etiquetas del índice de las columnas afectadas y el valor asociado a cada una de las claves, será el valor por defecto que sustituye a `NaN`.

````python
d = { 'precio': -100, 'Service' : 99}
hoteles.fillna(d)
````

#### 4.1.3.2 Reemplazo por el valor de la media

````python
# reemplaza por el valor de la media de la columna
hoteles['precio_2'] = hoteles['precio'].fillna(hoteles['precio'].mean())
hoteles.head()
````

#### 4.1.3.3 Forward / Backward fill (series temporales)

Muy útil en datos ordenados por fecha.

````python
#Propagación hacia adelante
hoteles['Service_ff'] = hoteles['Service'].ffill()  # usa valor anterior
hoteles.head()
````

````python
# Propagación hacia atrás
hoteles['Service_bf'] = hoteles['Service'].bfill()  # usa valor siguiente
hoteles.head()
````

### 4.1.4 Eliminar valores faltantes

Una alernativa a la imputación, es la eliminación de valores nules. La operación `dropna` eliminará todas las filas y/o todas las columnas que tengan algún valor NaN.

````python
# Eliminar filas con NaN
hoteles.dropna()
````

````python
# borra columnas que tengan aún NaN
hoteles.dropna(axis = 1)  
````

````python
# borra solo las filas que tengan NAN en la columna precio
hoteles.dropna(subset=['precio'])
````

### 4.3.3 Funciones map y apply

Son funciones para transformar datos.

#### 4.3.3.1 map - aplicar funciones a cada elemento

La función `map` permite aplicar una función a cada uno de los elementos de una serie (si se aplica a una serie) o un dataframe (si se aplica a un dataframe). 

````python
# hacemos un descuento del 10% a hoteles cuyo precio es mayor de 300
hoteles['descuento'] = hoteles['precio'].map(lambda x : x*0.1 if x > 100 else 0)
hoteles.head()
````

````python
# hacemos un descuento del 10% a hoteles cuyo precio es mayor de 300
# usamos una función definnida por el usuario
def calcular_descuento(valor):
    if valor > 100:
        return valor * 0.1
    else:
        return 0 

hoteles['descuento'] = hoteles['precio'].map(calcular_descuento)
hoteles.head()
````

#### 4.3.3.2 apply - Aplicar funciones a filas o columnas (serie completa)

La función `apply` permite aplicar una función a lo largo de uno de los ejes de un dataframe, que puede ser filas o columnas. Por tanto, la función aplicada con `apply` __recibe como argumento cada una de las filas (axis = 1) o cada una de las columnnas (axis = 0)__.

````python
# aplicamos un descuento solo a los hoteles del distrito Centro
def calcular_descuento(fila):
    if fila['distrito'] == 'Centro':
        return fila['precio'] * 0.1
    else:
        return 0 

hoteles['descuento_centro'] = hoteles.apply(lambda row: calcular_descuento(row), axis = 1)
hoteles.head()
````

# 5 Parte IV: Operaciones avanzadas

## 5.1 Grupos y operaciones de agregación (`groupby`)

Cuando queremos resumir información por categorías (por ejemplo, por `distrito`), usamos `groupby`.
Esto permite calcular métricas como conteo, media, máximo, mínimo o suma para cada grupo.

Estructura general:
- `df.groupby('columna')`
- y después aplicar agregaciones: `.mean()`, `.sum()`, `.count()` o `.agg({...})`

````python
# 1) Conteo de hoteles por stars
hoteles_por_estrellas = hoteles.groupby('stars')['code'].count()
hoteles_por_estrellas
````

````python
# 2) Agregaciones múltiples de precio por distrito
resumen_precios =  hoteles.groupby('distrito')['precio'].agg(['count', 'mean', 'min', 'max'])
resumen_precios.head()
````

````python
# 3) Agregaciones múltiples de precio por distrito
resumen_precios = (
    hoteles.groupby('distrito')['precio']
    .agg(['count', 'mean', 'min', 'max'])
    .rename(columns={
        'count': 'numero_hoteles',
        'mean': 'precio_medio',
        'min': 'precio_min',
        'max': 'precio_max'
    })
    .round(2)
    .sort_values('precio_medio', ascending=False)
 )

resumen_precios.head()
````

````python
# 4) Ejemplo con varias columnas (adaptable según columnas disponibles)
agg_dict = {'precio': ['mean', 'median'], 
            'stars': ['max']}

resumen_completo = hoteles.groupby('distrito').agg(agg_dict).round(2)
resumen_completo.head()
````

````python
# 5) Agrupamos por varias columnas
hoteles.groupby([ 'stars', 'distrito']).agg({'precio': ['mean', 'min', 'max']})
````

## 5.2 Pivot

Las funciones `.pivot()` y `.pivot_table()` se utilizan para reorganizar datos y construir tablas resumen tipo Excel (tablas dinámicas).

Son fundamentales en análisis exploratorio y reporting.

Se usa la sintaxis:

```python
   df.pivot_table(index=..., columns=..., values=..., aggfunc=...)
```

donde: 

* `index` - filas (pueden ser varias)

* `columns` - columnas nuevas

* `values` - valores a mostrar

* `aggfun` - pueden ser varias (`mean`, `min`, `max`, `std`, `count`, etc.)

````python
# Número de vuelos operados con cada origen-destino
hoteles.pivot_table(
    index='distrito',
    columns='stars',
    values='code',
    aggfunc='count'
).head(5)
````

### 5.2.1 Rellenar valores missing y totales generales

Otros argumentos:

*  `fill_value` para evita valores `NaN` en el resultado.
*  `margins` para añadir totales generales.
    

````python
hoteles.pivot_table(
    index='distrito',
    columns='stars',
    values='code',
    aggfunc='count',
    fill_value=0,
    margins=True
).head()
````

## 5.3 Merge: combinar DataFrames

La operación `merge` permite unir dos tablas usando una clave en común, igual que un `JOIN` en SQL.

En Pandas se usa con `pd.merge()` y el parámetro `how` para indicar el tipo de unión:
- `inner`: solo coincidencias en ambos DataFrames
- `left`: todas las filas de la tabla izquierda
- `right`: todas las filas de la tabla derecha
- `outer`: todas las filas de ambas tablas

````python
# Dos DataFrames pequeños inventados de hoteles
hoteles_info = pd.DataFrame({
    'hotel_id': ['H1', 'H2', 'H3', 'H4'],
    'nombre': ['Sol Centro', 'Gran Vía Inn', 'Retiro Suites', 'Plaza Norte'],
    'distrito': ['Centro', 'Centro', 'Retiro', 'Chamartín']
})

hoteles_resenas = pd.DataFrame({
    'hotel_id': ['H1', 'H2', 'H5'],
    'valoracion_media': [8.7, 9.1, 7.9],
    'num_resenas': [210, 340, 75]
})

print('hoteles_info')
display(hoteles_info)
print('hoteles_resenas')
display(hoteles_resenas)
````

````python
# inner: combina los dataframes devolviendo únicamente aquellas 
#        filas que tienen valores idénticos en las columnas que se comparan. 
inner_merge = pd.merge(hoteles_info, hoteles_resenas,
                       on='hotel_id',
                       how='inner')
inner_merge
````

````python
# `left` combina dos dataframes devolviendo aquellas filas que tienen valores idénticos
#        en las columnas que se comparan, y todas las filas del dataframe de la izquierda.
#        Las que no tengan correspondencia se rellenan con `NaN`.
left_merge = pd.merge(hoteles_info, hoteles_resenas,
                      on='hotel_id', 
                      how='left')
left_merge
````

````python
# right
right_merge = pd.merge(hoteles_info, hoteles_resenas, 
                       on='hotel_id', 
                       how='right')
right_merge
````

````python
outer_merge = pd.merge(hoteles_info, hoteles_resenas, on='hotel_id', how='outer', indicator=True)
outer_merge
````

### 5.3.1 Merge con claves de nombre distinto

Si la columna clave no se llama igual en ambos DataFrames, usamos `left_on` y `right_on`.

````python
# Mismo contenido, pero renombrando la clave en el DataFrame de reseñas
hoteles_resenas2 = hoteles_resenas.rename(columns={'hotel_id': 'codigo_hotel'})
hoteles_resenas2
````

````python

merge_columnas_distintas = pd.merge(
    hoteles_info,
    hoteles_resenas2,
    left_on='hotel_id',
    right_on='codigo_hotel',
    how='left'
)

merge_columnas_distintas
````

## 5.4 Concat: apilar o unir DataFrames

La operación `concat` se usa para combinar DataFrames por filas o por columnas.

Con `pd.concat()` es posible:
- **Apilar filas** (`axis=0`, opción por defecto)
- **Unir columnas** (`axis=1`)

Es útil cuando los datos vienen en varios ficheros/partes y es necesario juntarlos en una sola tabla.

````python
# Dos DataFrames pequeños  de hoteles
hoteles_parte1 = pd.DataFrame({
    'hotel_id': ['H1', 'H2'],
    'nombre': ['Sol Centro', 'Gran Vía Inn'],
    'distrito': ['Centro', 'Centro']
})

hoteles_parte2 = pd.DataFrame({
    'hotel_id': ['H3', 'H4'],
    'nombre': ['Retiro Suites', 'Plaza Norte'],
    'distrito': ['Retiro', 'Chamartín']
})

print('hoteles_parte1')
display(hoteles_parte1)
print('hoteles_parte2')
display(hoteles_parte2)
````

````python
# Concat por filas (axis=0): apilar registros
hoteles_todos = pd.concat([hoteles_parte1, hoteles_parte2] )
hoteles_todos
````

````python
# Concat por filas con nuevo índice continuo
hoteles_todos_idx = pd.concat([hoteles_parte1, hoteles_parte2],                         
                               ignore_index=True)
hoteles_todos_idx
````

### 5.4.1 Concat por columnas (`axis=1`)

Para concatenar por columnas, ambos DataFrames se alinean por índice.

````python
# DataFrame con métricas para los mismos hoteles  
metricas = pd.DataFrame({
    'precio_medio': [120, 155, 5],
    'valoracion_media': [8.7, 9.1, 6]
})

concat_columnas = pd.concat([hoteles_parte1, metricas], axis=1)
concat_columnas
````
