## 1.1 Módulos definidos por el usuario

En Python, cada fichero con extensión `.py` se considera un módulo. Los módulos en python pueden contener variables, funciones y clases.

__Ejemplo:__

Podemos crear un módulo llamado `funciones.py` que contenga tres funciones.
* factorial
* factoria_iterativo (solo para números grandes)
* factorial_recursivo (para números pequeños)

````python
# funciones_pk.funciones.funciones_itilities.py

def factorial_iterativo(n:int)->int:
    """
    Calcula el factorial de un número entero no negativo de forma iterativa.

    El factorial de un número ``n`` se define como el producto de todos los 
    enteros positivos desde 1 hasta ``n``. Por convención, el factorial de 0 es 1.

    :param n: Un número entero no negativo.
    :type n: int
    :return: El factorial del número dado.
    :rtype: int
    :raises ValueError: Si el número proporcionado es negativo.

    Examples
    --------
    >>> n = 4
    >>> factorial_iterativo(n)
    24
    >>> factorial_iterativo(0)
    1
    >>> factorial_iterativo(1)
    1
    >>> factorial_rec(-3)
    Traceback (most recent call last):
      ...
    ValueError: El número debe ser un entero no negativo.
    """
    if n < 0:
        raise ValueError("El número debe ser un entero no negativo.")
    resultado = 1
    for i in range(2, n + 1):
        resultado *= i
    return resultado


def factorial_rec(n:int)-> int:
    """
    Calcula el factorial de un número entero no negativo de forma recursiva.

    El factorial de un número ``n`` se define como el producto de todos los 
    enteros positivos desde 1 hasta ``n``. Por convención, el factorial de 0 es 1.

    :param n: Un número entero no negativo.
    :type n: int
    :return: El factorial del número dado.
    :rtype: int
    :raises ValueError: Si el número proporcionado es negativo.

    Examples
    --------
    >>> factorial_rec(6)
    720
    >>> factorial_rec(0)
    1
    >>> factorial_rec(1)
    1
    >>> factorial_rec(-4)
    Traceback (most recent call last):
      ...
    ValueError: El número debe ser un entero no negativo.
    """
    if n < 0:
        raise ValueError("El número debe ser un entero no negativo.")
    if n == 0 or n == 1:
        return 1
    return n * factorial_rec(n - 1)

def factorial(n:int)-> int:
    """
    Calcula el factorial de un número entero no negativo. El cálculo se hará 
    de forma iterativa o recursiva dependiendo del valor del número.

    El factorial de un número ``n`` se define como el producto de todos los 
    enteros positivos desde 1 hasta ``n``. Por convención, el factorial de 0 es 1.

    :param n: Un número entero no negativo.
    :type n: int
    :return: El factorial del número dado.
    :rtype: int
    :raises ValueError: Si el número proporcionado es negativo.

    Examples
    --------
    >>> r = factorial(1100)
    >>> str(r)[0:26] == '53437084880926377034242155'
    True
    >>> factorial(0)
    1
    >>> factorial(1)
    1
    >>> factorial(-4)
    Traceback (most recent call last):
      ...
    ValueError: El número debe ser un entero no negativo.
    """
    if n > 1000:
        return factorial_iterativo(n)
    else:
        return factorial_rec(n)
````

### 1.2.1 El fichero `__init__.py`

Si una carpeta contiene un archivo `__init.py__`, Python lo trata como un paquete, permitiendo que otros módulos puedan importarlo. No es obligatorio incluir este archivo, pero es útil para configurar la interfaz del paquete.

* El `__init__.py` puede estar vacío. Define la interfaz pública del paquete y oculta la complejidad interna de las carpetas y archivos.
* La variable `__all__` indica qué módulos o subpaquetes se incluye si se hace `from package import *`. Su función es indicar qué elementos del paquete forman parte de la API pública. Si no se define `__all__`, la importación con asterisco (*) importará todos los nombres que no comiencen con un guion bajo.
* Dentro de `__init__.py` debemos usar importaciones relativas (como `from .modulo1 import ...`) para evitar problemas de resolución de nombres.

En el ejemplo de la imagen, mostramos el aspecto del archivo `__init__.py` del paquete `polinomials`. 

* La clase `Monomial()` está definida en el módulo `mono.py`

### 1.2.2 Acceder a ficheros del paquete

Acceder a los ficheros del paquete puede parecer fácil, pero
debemos entender que cuando se instalen no podremos saber
dónde se instalarán. Para ello vamos a necesitar el paquete
`importlib.resources`.

````python
import importlib.resources
from store import *
import pandas as pd


def read_data(filename: str|None = None) -> pd.DataFrame:
    """
    This function reads a file with the required format. It returns store dictionary
    >>> read_data(importlib.resources.files('store').joinpath('data/store.txt')).columns
    Index(['cod', 'loc', 'lat', 'stock', 'cod_prov', 'name', 'price', 'val1',
           'val2'],
          dtype='str')

    >>> read_data().columns
    Index(['cod', 'loc', 'lat', 'stock', 'cod_prov', 'name', 'price', 'val1',
           'val2'],
          dtype='str')
    """
    if filename == None:
        filename = importlib.resources.files('store').joinpath('data/store.txt')
    df = pd.read_csv(filename, sep = '//', engine='python')
    
    return df
    
````

## 1.3 Automatización de paquetes (fichero `toml`)

La instalación de paquetes se puede automatizar de forma sencilla y se pueden dar de alta en PyPi (https://pypi.org/) para que otros usuarios puedan usarlo.

Es necesario generar un fichero llamado `pyproject.toml` que incluye instrucciones para generar el fichero `wheel` (que es el instalador de python). Este archivo define cómo se construye y distribuye un paquete, qué dependencias necesita, y más.

Este fichero `pyproject.toml` no está dentro del paquete, pero sí dentro del proyecto.

__Descripción de las secciones:__

* __[build-system]__:  Especifica los requerimientos necesarios para construir el paquete.  La diferencia principal en la sección [build-system] radica en la herramienta elegida para empaquetar el proyecto:

  - **Izquierda**: se utiliza `setuptools` (el motor de construcción tradicional y estándar de la comunidad Python). En este caso, usa `setuptools` como herramienta de construcción y define `setuptools.build_meta` como la interfaz.
    
  - **Derecha**: se usa `uv_build`, el backend nativo del ecosistema de `uv`. En la práctica, `uv_build` optimiza este proceso integrándolo de forma nativa con el resto de comandos de `uv`.

  
* __[project]__:

    * `name`: Nombre del paquete.
    * `version`: Versión del paquete.
    * `requires-python`: Especifica la versión mínima de Python compatible.
    * `authors`: Lista de autores del paquete, con nombre y correo.
    * `dependencies`: Lista de dependencias requeridas. Se incluyen las dependencias de paquetes externos a Python. Cuando se instale el paquete, se comprueba que estén instaladas. Si no lo están, se descargan de Pypi. Ejemplo:
      
      ```
              dependencies = [ " requests ~= 2.1 ", " pandas ~= 2.1 " ]
      
      ```


  Los especificadores se escriben normalmente con la sintaxis de la especificación de dependencias de Python.

  Ejemplos:

  | Especificador | Significado aproximado |
  |---|---|
  | `pandas==2.2.3` | Instala únicamente la versión 2.2.0.|
  | `pandas>=2.0` |  Instala la 2.0.0 o cualquiera superior |
  | `pandas<3` | Evita instalar versiones mayores a la especificada. |
  | `pandas>=2,<3` | Rango cerrado. Instala cualquier versión entre la 2 y la 3 (excluyendo la 3). |
  | `pandas!=2.1.0` | Excluye una versión 2.1.0 |
  | `pandas ~= 2.1.0` |(Operador de compatibilidad): Equivalente a >= 2.1.0, == 2.1.*. Permite actualizaciones de parches dentro de la misma versión menor, pero evita saltar a la 2.2.0. |


* **[dependency-groups]**: Es el grupo de dependencias de desarrollo. Con uv se  separa limpiamente las herramientas de desarrollo y testing (pytest, coverage, pytest-cov, pytest-datafiles) del resto del proyecto.   
Estas herramientas no se incluirán como requisitos al instalar el paquete generado (.wheel)


* __[tool.setuptools.packages.find]__:  Configura cómo setuptools debe localizar los paquetes dentro del proyecto.

  
    * **exclude**: Lista de patrones de nombres que no se deben incluir. Se usa para expluir la carpeta de tests.
<br/>
   * En `uv_build`, la exclusión se configura mediante **[tool.uv.build]** . Sin embargo, si usamos la estructura estándar `src/`, la exclusión de tests y archivos externos ocurre de forma automática sin necesidad de especificar la sección **[tool.uv.build]**.    Para exluir archivos o carpetas que se encuentran dentro de `src/`:
   ```python
            [tool.uv.build]
            exclude = [
                "src/store/**",
            ]
   ```


* __[tool.setuptools.package-data]__:     Define qué datos adicionales deben incluirse en el paquete al distribuirlo. En este caso, archivos `.txt` y `.csv` en la carpeta `data`. en el caso de `uv_build`, los datos ubicados dentro de la carpeta del paquete (src/) se incluyen automáticamente sin necesidad de configurarlo.


## 1.4 Construcción del paquete con  uv


### Con uv

Desde la carpeta donde se encuentte en archivo `pyproject.toml`, abrir una terminal y escribir:

```python   
    uv build   
```



### Sin uv

* Es necesario instalar el paquete `build` (pip install build)
* Desde la carpeta donde se encuentte en archivo `pyproject.toml`, abrir una terminal y escribir


## 1.5 Instalar el paquete

Para probar la instalación del archivo .whl recién construido en otro proyecto:

### Con uv

Lo instalamos:

```python
uv add ruta\ejemplo_paquete\dist\ejemplo_paquete-0.1.0-py3-none-any.whl
```
Si necesitamos cambiar el código fuente del paquete sin cambiar la versión en el toml, tenemos que volver a instalarlo. Lo mejor es usar:

 ```python 
 uv pip install ruta\ejemplo_paquete\dist\ejemplo_paquete-0.1.0-py3-none-any.whl --reinstall 
 ```
Esto solo lo mete a la fuerza en tu entorno virtual sin actualizar el archivo de bloqueo `uv.lock`.



### Sin uv 

Para instalarlo con pip desde la terminal:

```python
pip install ruta\ejemplo_paquete\dist\ejemplo_paquete-0.1.0-py3-none-any.whl --reinstall 

```
