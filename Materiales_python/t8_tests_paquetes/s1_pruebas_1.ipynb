# Pruebas con Doctest

---

Cuando se desarrollan paquetes, es conveniente probar el codigo para asegurar en la medida de lo posible su correcto funcionamiento. Para ello se suelen utilizar distintos tipos de pruebas. Aquí nos centraremos en las pruebas unitarias, mediante las que se comprueba el correcto funcionamiento de las unidades lógicas en las que se divide un programa, sin tener en cuenta la interrelación con otras unidades.

Para las pruebas unitarias vamos a usar `pytest`, combinado con `doctest` para pruebas más sencillas. Solo `doctest`  está incluído en la librería estándar de Python.


### Doctest

__Doctest__ permite combinar las pruebas con la documentación. Esta idea de utilizar las pruebas unitarias para probar el código y también a modo de documentación tiene muchos beneficios:
* permite realizar pruebas de forma muy sencilla,
* propicia el que las pruebas se mantengan actualizadas,
* sirve a modo de ejemplo de uso del código y como ayuda para entender su propósito.

Cuando `doctest` encuentra una línea en la documentación que comienza con >>> se asume que lo que le sigue es código Python a ejecutar, y que la respuesta esperada se encuentra en la línea o líneas siguientes, sin >>>. El texto de la prueba termina cuando se encuentra
una línea en blanco, o cuando se llega al final de la cadena de documentación.

Veamos un ejemplo:

````python
def gcd(a:int, b:int)-> int:
    """
    Calcula el máximo común divisor (MCD) de dos números enteros utilizando el algoritmo de Euclides.
    :param a: El primer número entero.
    :param b: El segundo número entero.    
    :returns: El máximo común divisor de `a` y `b`.    
    :raises ValueError: Si cualquiera de los dos números es negativo.
    :example:
    >>> a = 1
    >>> gcd(a, 2)
    1
    >>> gcd(48, 18)
    6
    >>> gcd(54, 24)
    6
    >>> gcd(7,0)
    Traceback (most recent call last):
      ...
    ValueError: (False, 'b must be > 0')
    """
    if a <0:
        raise ValueError ('a must be >= 0')
    if b <=0:
        raise ValueError(b > 0, 'b must be > 0')
    while b !=0:
        r = a % b
        a = b
        b = r
    return a 
````

El primer ejemplo escrito en el docstring de la función, indicamos a doctest que cree un variable `a` con valor 1, que llame a continuación a la función `gcd` con los argumentos `a` y 2, y que compruebe que el resultado devuelto sea igual a 1.

__Ejecución de las pruebas:__

Para ejecutar las pruebas se utiliza la función `testmod` del módulo `doctest`, a la que se le puede pasar opcionalmente el nombre de un módulo a evaluar (parámetro name). En el caso de que no se indique ningún argumento, como en este caso, se evalúa el módulo actual.

````python
import doctest
def test():   
    doctest.testmod(verbose= True)
````

````python
test()
````

````python
if __name__ == '__main__':
    test()
````

Ahora vamos a introducir un error en el código de la función para ver el aspecto de un mensaje de error de doctest. Supongamos, por
ejemplo, que hubieramos escrito un operador de división ('/') en lugar de módulo ('%')

````python
def gcd(a:int, b:int)-> int:
    """
    Calcula el máximo común divisor (MCD) de dos números enteros utilizando el algoritmo de Euclides.
    ...
    :example:
    >>> a = 1
    >>> gcd(a, 2)
    1
    >>> gcd(48, 18)
    6
    >>> gcd(54, 24)
    6
    >>> gcd(7,0)
    Traceback (most recent call last):
      ...
    ValueError: (False, 'b must be > 0')
    """
    if a <0:
        raise ValueError ('a must be >= 0')
    if b <=0:
        raise ValueError(b > 0, 'b must be > 0')
    while b !=0:
        r = a / b    # introducimos aquí un error para que fallen los test
        a = b
        b = r
    return a 
````

````python
test()
````

Como vemos, el mensaje nos indica que ha fallado la prueba de la línea 7.

### Sentencias anidadas

Veamos por último cómo utilizar sentencias anidadas para hacer cosas un poco más complicadas con doctest. En el ejemplo siguiente diseñamos una prueba que compruebe que el resultado es el adecuado para varias llamadas con distintos valores. Las sentencias
anidadas comienzan con '...'  en lugar de >>>.

````python
def gcd(a:int, b:int)-> int:
    """
    Calcula el máximo común divisor (MCD) de dos números enteros utilizando el algoritmo de Euclides.
    Aquí los parámetros
    :example:
    >>> for a, b in [(1,2), (48,18), (54, 24)]:
    ...     gcd(a, b)
    1
    6
    6
    """
    if a <0:
        raise ValueError ('a must be >= 0')
    if b <=0:
        raise ValueError(b > 0, 'b must be > 0')
    while b !=0:
        r = a % b
        a = b
        b = r
    return a 
````

## Ejecución de los doctest con cobertura


Para medir la cobertura de las pruebas creadas con doctest es puede usar la  librería `coverage`

```python
    uv add --dev coverage 
```



| Objetivo de Ejecución | Comando con `uv` | Notas / Descripción |
| :--- | :--- | :--- |
| **Doctest directo + Coverage (Archivo único)** | `uv run coverage run -m doctest archivo.py` | Ejecuta las pruebas integradas en los *docstrings* del archivo usando la herramienta nativa `doctest`. |
| **Ver reporte de Coverage en consola** | `uv run coverage report` | Muestra el porcentaje de líneas probadas en la terminal tras haber ejecutado `coverage run`. |
| **Generar reporte HTML de Coverage** | `uv run coverage html` | Genera la carpeta `htmlcov/` con un reporte visual interactivo por archivo y línea. |

En PyCharm Professional se incluye un  analizador de cobertura visual (sin necesidad del comando coverage de terminal):

* Selecciona Run 'Doctest in...' with Coverage (o la opción equivalente de pytest si tienes configurado pytest como ejecutor predeterminado).

PyCharm ejecutará los tests y resaltará en verde las líneas probadas y en rojo las no probadas directamente sobre la barra de desplazamiento del editor de código.
