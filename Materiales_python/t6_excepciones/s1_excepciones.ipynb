## ¿Qué es una excepción en programación?

Las excepciones son un tipo de error que  aparecen durante la ejecución del programa.
* Pueden producirse por un uso incorrecto del programa por parte del usuario, por ejemplo si el usuario introduce una cadena cuando se espera un número. 
* En otras ocasiones pueden deberse a errores de programación, por ejemplo si una función intenta acceder a la quinta posición de una lista de 3 elementos o realizar una división por cero.

## Captura y tratamiento de las excepciones

* Si no se captura la excepción, el programa principal acaba dando un error
* Las excepciones se pueden capturar para que el programa principal no lance ese error
* No siempre es bueno capturar las excepciones. A veces es recomendable que el programa se rompa para poder encontrar el problema

Para tratar los errores de ejecución es necesario capturar las excepciones en nuestros programas mediante las instrucciones __try / except__.

1.  `try ... except`
2.  `try ... except ... else ` 
3.  `try ... except ... else ... finally` 
4.  `try ... except ... finally` 
5.  `try ... finally` 

Todo es opcional, debe haber un `finally` o un `except`.

Ejemplo con sintaxis completa:

```python
try:
    # Código que podría generar una excepción
    ...
except TipoDeExcepcion:
    # Código que se ejecuta si ocurre esa excepción 
    # podemos tener varios bloques except, una por tipo de excepción que quweremoa tratar
    ...
except:
    # Código que se ejecuta si ocurre cualquier excepción 
    ...
else:
    # Opcional: se ejecuta si NO hubo excepción
    ...
finally:
    # Opcional: se ejecuta SIEMPRE, haya o no excepción
```


### Sentencias `try ... except`

Primero, se ejecuta el bloque `try`. 
* Si no ocurre ninguna excepción, los bloques `except` se saltan y termina la ejecución de la sentencia `try`.
* Si ocurre una excepción durante la ejecución del bloque `try`, el resto de dicho bloque se salta. Si la excepción que ha saltado coincide con la indicada en la cláusula `except`, se ejecuta el bloque `except`, y la ejecución continúa con la instrucciones que haya fuera de la sentencia `try`.

__Ejemplos:__

````python
b = 10
try:
    a = int(input('Introduce un número: '))
    print(b / a)
except:
    print('Has introducido 0.')
````

````python
lista = [1, "dos", 3.0, "cuatro", 5,   6.3]


for e in lista:
    try:
        print(float(e))
    except ValueError:       # capturamos sola las  excepciones de tipo ValueError 
        print(e,"error")    
````

También se puede capturar varias excepciones con la sentencia `try ... except`.  En el caso anterior ha saltado una excepción del tipo __TypeError__. Para tratarla debemos añadir otra captura en nuestro código.

````python
lista = [ "1", (12, 14), "dos", 3.0, "cuatro", 5 ]
for e in lista:
    try:
        print(f'Resultado: {e} - {float(e)}')
    except ValueError:       # capturamos la excepción
        print(e,"error de tipo ValueError")
    except TypeError:
        print(e, "esto no me lo esperaba")
    except:
        print( 'esto trata todas las excepiones lno controlada anteriormente')
````

````python
lista = [ 1, "s", 3.0, 5 ]
for e in lista:
    try:
        print(float(e))
    finally:
        print("Fin")
    
````

El bucle anterior solo se ejecuta dos veces. La segunda produce error.

### Bloque `try ... except ... else`

Las declaraciones `try ... except` tienen un bloque `else` opcional, el cual, cuando está presente, debe seguir a los `except`. Es útil para aquel código que debe ejecutarse si el bloque `try` no genera una excepción. 

__Ejemplo:__

Supongamos que tenemos una lista con nombres de ficheros. Si intentamos abrir cada uno de ellos mediante la instrucción `open`, se lanzará una excepción si el fichero no existe.

````python
list_files = ['datos/cuento.txt', 'fuuuuuuu.txt']

for file in list_files:
    try:
        f = open(file, 'r')
        print('El fichero ', file, 'tiene', len(f.readlines()), 'lineas')
    except KeyError:
        print('Error de clave')
    except Exception as e: 
        print(f'Se captura error al abrir {file}')
        print(f'Tipo de excepción lanzada: {type(e).__name__}')
        print(f'tipo {type(e)}')
    else:   
        print('Cerramos fichero')     
        f.close()
````

Si se quiere hacer algo con la excepción, se debe declarar con `as excep`. (excep es una variable que contiene la excepción producida.)

### Bloque try ... except ... finally

El bloque `finally` opcional, el cual, cuando está presente, debe ser el último. La cláusula `finally` es muy útil porque siempre se ejecuta, independientemente de si ocurrió una excepción o no.

Finally garantiza que cierto código se ejecute sin importar lo que pase en el bloque try. Esto es especialmente importante para:

* Liberar recursos (cerrar archivos, conexiones a bases de datos, sockets, etc.).
* Limpiar el estado del programa.
* Registrar eventos (por ejemplo, logs de finalización).

Ejemplo:

````python
archivo = None
try:
    archivo = open('datos/cuento.txt', "r")
    contenido = archivo.readlines()
    print(f'El fichero tiene {len(contenido)} líneas')
except FileNotFoundError:
    print("El archivo no existe.")
finally:
    if archivo and not archivo.closed:
        archivo.close()
        print("Archivo cerrado correctamente.")
````

Un ejemplo con todo:

````python
def f1(l: list) -> None:
    l[0] = l[0] + 1

def f2(l: list) -> int:
    f1(l)
    return l[0]/l[1]

def f3(l):
    if l[1] == 1:
        raise ValueError('l[1] == 1')

def main():
    m = [
        [],
        [1, 0],
        [1, 1],
        [1, 2]
    ]
    for l in m:
        try:
            print(f'{l}....', end='')
            f2(l)
        except IndexError as e:
            print(f"Salta excepción: Error de rango: '{e}'")
        except ZeroDivisionError:
            print(f"Salta excepción: División por cero'")
        else:
            print("Esto se ejecuta si no se la lanado excepción")
        finally:
            print("Esto se ejecuta siempre")
            print('-'*25)

main()
````

## Lanzamiento de excepciones

El mecanismo de las excepciones también sirve para que nuestros programas indiquen que se ha producido un error, por ejemplo
cuando los datos de entrada de una función no son los adecuados. En estos casos usamos __raise__. De esta forma, otros programas que estén usando nuestro código pueden responder a ese error.

Las excepciones que podemos lanzar por ahora son aquellas que están integradas en Python y que se encuentran definidas en el módulo `exceptions`. Nunca es necesario importar este módulo explícitamente.

Agunas de las excepciones integradas son:

| ZeroDivisionError | Se lanza cuando el segundo argumento de una operación de división o módulo es cero| |
| IOError  | Se lanza cuando una operación de E/S (la función integrada open()) fracasa por motivos relativos a E/S  |   |  
| IndexError | Se lanza cuando un sub-índice de una secuencia se sale del rango | |
| KeyError | Se lanza cuando no se encuentra una clave de una correspondencia (diccionario) | |
| SyntaxError |Se lanza cuando aparecen errores de sintaxis |IndentationError|
| IndentationError |Se lanza cuando una indentación incorrecta sucede||
| NameError | Lanzada cuando no se encuentra un nombre local o global | |

En el siguiente ejemplo, la función lanzará una excepción de tipo `ValueError` en el caso de que el parámetro de entrada `radius` sea negativo:

````python
import math
def circle(radius: float)-> float:
    """
    Function that computes the surface of a circle
    :param radius: real
    :returns: rtype: real

    :raises: ValueError cuando radius es negativo
    :precondition: radius>=0

    :example
    >>> circle(3)
    12.566370614359172   
    """
    if radius >= 0:
        sur = math.pi * radius ** 2
    else:
        raise ValueError('Radio negativo.... !!!!!!!')
    return sur
````

### Información asociada a las excepciones

Normalmente las excepciones tienen un valor asociado indicando en detalle la causa del error. 
* Este valor puede ser una cadena o tupla de varios elementos informativos (es decir, un código de error y una cadena que explica el código). 


__Ejemplo:__

Supongamos que queremos lanzar una excepción con la información de todos los valores mayores que 1000:

````python
def validar(Lista):
    valores_erroneos = []
    for valor in Lista:
        if valor > 1000:
            valores_erroneos.append(valor)
    if valores_erroneos != []:
        raise ValueError(tuple(valores_erroneos))   # lanzamos una excepción con una tupla
    return True 
````

# Excepciones definidas por el usuario

Los usuarios pueden lanzar sus propias excepciones, igual que pueden crear sus propias funciones o clases. 

Para ello es necesario crear una nueva clase excepción. Las clases excepción creadas por el usuario, deberán derivar de la clase `Exception` o de alguna de sus derivadas.

Las clases excepción pueden ser definidas de la misma forma que cualquier otra clase.

__Ejemplo:__


````python
class ErrorPunto(Exception):
    """Ejemplo de Clase exception para excepciones lanzadas por la clase Punto.
       Pero cualquier función puede lanzar este tipo de excepciones.
    """
    pass
````

Supongamos que creamos una clase `Point`  para representar puntos en el plano que no estén situados en los ejes de coordenadas. Es decir, un punto es válido si `x != 0 ` e `y != 0`. 
El método constructor `__init__` deberá lanzar una excepción en el caso de que los argumentos de entrada no cumplan con las condiciones.

````python
class Point:
    def __init__(self , cx , cy  ):
        if cx == 0 or cy == 0:
            raise ErrorPunto((cx, cy))
        else:
            self._x = cx
            self._y = cy  
````

````python
p = Point(6, 5)     # crea el punto correctamente
````

````python
p = Point(6, 0)     # se produce un error en la creación del punto
````

Como podemos ver, el tipo de excepción que salta, es la excepción de usuario `ErrorPunto`.

````python
try:
    p = Point(6, 0)  
except ErrorPunto as e:
    print('No se puede crear un punto con las coordenadas: ', e)
````

__Ejemplo:__

Pero las clases de tipo exception también pueden tener sus propios atributos y métodos.

````python
class EntradaErrorPunto(Exception):
    """Excepción lanzada por errores en las entradas.

    Atributos:        
        mensaje -- explicación del error
        valor_x -- valor incorrecto
        valor_y -- valor incorrecto
    """

    def __init__(self, mensaje, x, y):
        self.mensaje = mensaje
        self.valor_x = x
        self.valor_y = y
````

````python
class Point:
    def __init__(self , cx , cy  ):
        if cx == 0:
            # Se lanza una excepción con mensaje apropiado
            raise EntradaErrorPunto('El valor es incorrecto para la coordenada x:', cx, cy)
        if cy == 0:
            raise EntradaErrorPunto('El valor es incorrecto para la coordenada y:', cx, cy)
        else:
            self._x = cx
            self._y = cy  
````

````python
p = Point(2, 3)
````

````python
p = Point(2, 0)
````

````python
p = Point(0, 6)
````

Si manejamos la excepción mediante `except`, es posible consultar sus atributos:

````python
try:
    p = Point(0, 8)  
except EntradaErrorPunto as e:
    print(e.mensaje)
   
````
